
### Step 1: Understand the Concept

**What is Feature Importance?**

Feature Importance is a technique used to measure how much each input feature contributes to a Machine Learning model's predictions.

It helps us understand which features are more useful for making predictions.

**What is a Feature?**

A feature is an input variable used by a Machine Learning model to make predictions.

Example:
- Customer_Age
- Annual_Income
- Purchase_Frequency

**What is Feature Importance Score?**

A Feature Importance Score represents the relative contribution of an input feature according to a particular model or importance calculation method.

- Higher score: The feature has greater importance according to the method.
- Lower score: The feature has lower importance according to the method.
- Zero score: The method assigns no importance to that feature.

**Why is Feature Importance Used?**

- Understand which features influence model predictions.
- Identify useful input variables.
- Improve model interpretability.
- Support feature selection decisions.
- Help communicate model behaviour to business teams.

**How is Feature Importance Calculated?**

Feature importance can be calculated using different approaches:

1. Model-Based Feature Importance
2. Decision Tree Feature Importance
3. Random Forest Feature Importance
4. Permutation Importance

Each method measures importance differently.

**Feature Importance in AI/ML**

Feature importance helps AI/ML Engineers understand how a trained model uses input features to generate predictions.

**Advantages:**

- Helps understand model behaviour.
- Identifies influential input features.
- Supports model interpretation.
- Helps investigate unnecessary or less useful features.

**Limitations:**

- Different methods can produce different importance rankings.
- Correlated features can affect importance scores.
- Importance does not prove causation.
- A high importance score does not guarantee better prediction performance.



### Step 2: Demonstrate the Concept

**Real-World Example: Customer Spending Prediction**

An online shopping company wants to predict customer spending.

The company uses the following input features:

- Customer_Age: Age of the customer.
- Annual_Income: Customer's yearly income.
- Purchase_Frequency: Number of purchases made by the customer.
- Customer_Visits: Number of times the customer visits the website.

The target variable is:

- Customer_Spending: Total amount spent by the customer.

**Understanding Feature Importance:**

Suppose a trained model produces the following illustrative importance scores:

| Feature | Importance Score |
|---|---:|
| Annual_Income | 0.45 |
| Purchase_Frequency | 0.30 |
| Customer_Visits | 0.15 |
| Customer_Age | 0.10 |

**Interpretation:**

- Annual_Income has the highest importance score.
- Purchase_Frequency is the second most important feature.
- Customer_Visits has a lower importance score.
- Customer_Age has the lowest importance score.

**Important Understanding:**

These scores represent the model's relative use of the features. They do not prove that Annual_Income causes higher Customer_Spending.

**AI/ML Application:**

Feature Importance helps an AI/ML Engineer understand which input variables contribute most to a customer spending prediction model.


```python

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor

# Create a sample customer dataset
df = pd.DataFrame({
    "Customer_Age": [22, 25, 28, 32, 35, 38, 42, 45,
                     24, 30, 34, 40],
    "Annual_Income": [20000, 25000, 30000, 40000,
                      45000, 55000, 65000, 75000,
                      22000, 35000, 48000, 70000],
    "Purchase_Frequency": [2, 3, 4, 5, 6, 7, 8, 9,
                           2, 4, 6, 8],
    "Customer_Visits": [3, 4, 5, 6, 7, 8, 9, 10,
                        4, 5, 7, 9],
    "Customer_Spending": [3000, 4000, 5000, 7000,
                          9000, 11000, 14000, 17000,
                          3500, 6000, 10000, 16000]
})

# Separate input features and target
X = df.drop(columns=["Customer_Spending"])
y = df["Customer_Spending"]

# Create the Random Forest model
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

# Train the model
model.fit(X, y)

# Calculate feature importance
importance_scores = model.feature_importances_

# Create a feature importance DataFrame
importance_df = pd.DataFrame({
    "Feature": X.columns,
    "Importance": importance_scores
})

# Sort features by importance
importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

print("Feature Importance Results:")
print(importance_df.round(4))

# Visualize feature importance
plt.figure(figsize=(8, 5))

plt.barh(
    importance_df["Feature"],
    importance_df["Importance"]
)

plt.xlabel("Importance Score")
plt.ylabel("Features")
plt.title("Random Forest Feature Importance")
plt.gca().invert_yaxis()

plt.show()
```

### Step 1: Understand the Concept

**Model-Based Feature Importance**

Model-Based Feature Importance is a technique used to identify the importance of input features based on the internal learning process of a trained machine learning model.

Different machine learning models calculate feature importance in different ways.

**Common Model-Based Methods:**

1. **Coefficient-Based Importance**
   - Used in Linear Regression and Logistic Regression.
   - Model coefficients indicate how input features are associated with the prediction.
   - Larger absolute coefficient values may indicate stronger influence when features are on comparable scales.

2. **Tree-Based Importance**
   - Used in Decision Tree and Random Forest models.
   - Importance is calculated based on how much a feature helps improve the model's predictions through splits.

**Why is Model-Based Feature Importance Used?**

- To understand which features the model relies on.
- To identify useful input features.
- To understand the model's prediction behavior.
- To support feature selection decisions.

**Important Note:**

Model-based importance depends on the selected algorithm. A feature may receive different importance scores from different models.

Feature importance does not prove that a feature causes the target variable to change.



### Step 2: Demonstrate the Concept

**Real-World Example: Customer Spending Prediction**

Imagine an online shopping company wants to predict how much a customer will spend.

The company collects the following information:

| Feature | Description |
|---|---|
| Customer_Age | Age of the customer |
| Annual_Income | Customer's yearly income |
| Purchase_Frequency | Number of purchases |
| Customer_Visits | Number of visits to the website |
| Customer_Spending | Total customer spending |

A Linear Regression model is trained using these features.

After training, the model produces coefficients for each input feature.

For example:

| Feature | Coefficient |
|---|---:|
| Customer_Age | 100 |
| Annual_Income | 0.5 |
| Purchase_Frequency | 800 |
| Customer_Visits | 200 |

These coefficients describe the relationship learned by the model.

However, their magnitudes cannot be directly compared here because the features use different units and scales.

After standardizing the input features, absolute coefficient magnitudes can be compared more meaningfully.

**Business Understanding:**

The company can use model-based importance to understand which customer details are useful for predicting spending.

The scores describe the model's learned relationships, not proof of cause and effect.


```python

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# Step 1: Create sample customer data
data = {
    "Customer_Age": [22, 25, 28, 32, 35, 38, 42, 45, 24, 30, 34, 40],
    "Annual_Income": [20000, 25000, 30000, 40000, 45000, 55000,
                      65000, 75000, 22000, 35000, 48000, 70000],
    "Purchase_Frequency": [2, 3, 4, 5, 6, 7, 8, 9, 2, 4, 6, 8],
    "Customer_Visits": [3, 4, 5, 6, 7, 8, 9, 10, 4, 5, 7, 9],
    "Customer_Spending": [3000, 4000, 5000, 7000, 9000, 11000,
                          14000, 17000, 3500, 6000, 10000, 16000]
}

df = pd.DataFrame(data)

# Step 2: Separate input features and target
X = df.drop("Customer_Spending", axis=1)
y = df["Customer_Spending"]

# Step 3: Create a pipeline with scaling and Linear Regression
model = Pipeline([
    ("scaler", StandardScaler()),
    ("regressor", LinearRegression())
])

# Step 4: Train the model
model.fit(X, y)

# Step 5: Extract model coefficients
coefficients = model.named_steps["regressor"].coef_

# Step 6: Create feature importance table
importance_df = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": coefficients,
    "Importance": abs(coefficients)
})

# Step 7: Sort features by importance
importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

print(importance_df)

# Step 8: Visualize feature importance
plt.figure(figsize=(9, 5))

plt.barh(
    importance_df["Feature"],
    importance_df["Importance"]
)

plt.xlabel("Absolute Standardized Coefficient")
plt.ylabel("Features")
plt.title("Model-Based Feature Importance")

plt.gca().invert_yaxis()
plt.show()
```

Selects the target column.

The pipeline performs two operations:

- StandardScaler transforms input features to a comparable scale.
- LinearRegression learns the relationship between the features and target.

Scaling is important because the input columns have different units.

**5. Train the Model**

`model.fit(X, y)`

The model learns relationships between customer information and spending.

**6. Extract Coefficients**

`model.named_steps["regressor"].coef_`

Retrieves the learned coefficients from Linear Regression.

A positive coefficient indicates a positive learned relationship, while a negative coefficient indicates a negative learned relationship, assuming other features remain constant.

**7. Calculate Importance**

`abs(coefficients)`

Takes the absolute value of each coefficient.

This allows us to compare coefficient magnitudes without considering their direction.

**8. Create and Sort the Table**

The DataFrame displays each feature, its coefficient, and its absolute coefficient value.

The features are sorted from highest to lowest importance.

**9. Visualize Importance**

The horizontal bar chart displays the relative coefficient magnitudes.

**Expected Output:**

The exact coefficient values are calculated when the notebook runs.

### Step 1: Understand the Concept

**Decision Tree Feature Importance**

Decision Tree Feature Importance is a technique used to measure how much each input feature contributes to the prediction process of a Decision Tree model.

A Decision Tree makes predictions by splitting data based on feature values.

During each split, the model tries to reduce impurity and create more meaningful groups.

**What is Impurity?**

Impurity measures how mixed or inconsistent the target values are within a group.

For regression problems, Decision Trees commonly use Mean Squared Error (MSE) or variance reduction.

For classification problems, common impurity measures include Gini Impurity and Entropy.

**How Feature Importance Works:**

1. The Decision Tree selects a feature for splitting.
2. The split divides the data into smaller groups.
3. The model measures how much impurity is reduced.
4. The reduction is accumulated for the feature across the tree.
5. The accumulated importance values are normalized.

**Important Points:**

- Features that produce larger impurity reductions generally receive higher importance.
- Features that are not used for splitting may receive zero importance.
- Importance values are specific to the trained model and dataset.
- Feature importance does not prove a causal relationship.

1. The model checks different features for possible splits.

```python

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeRegressor

# Step 1: Create sample customer data
data = {
    "Customer_Age": [22, 25, 28, 32, 35, 38, 42, 45, 24, 30, 34, 40],
    "Annual_Income": [20000, 25000, 30000, 40000, 45000, 55000,
                      65000, 75000, 22000, 35000, 48000, 70000],
    "Purchase_Frequency": [2, 3, 4, 5, 6, 7, 8, 9, 2, 4, 6, 8],
    "Customer_Visits": [3, 4, 5, 6, 7, 8, 9, 10, 4, 5, 7, 9],
    "Customer_Spending": [3000, 4000, 5000, 7000, 9000, 11000,
                          14000, 17000, 3500, 6000, 10000, 16000]
}

df = pd.DataFrame(data)

# Step 2: Separate input features and target
X = df.drop("Customer_Spending", axis=1)
y = df["Customer_Spending"]

# Step 3: Create a Decision Tree model
model = DecisionTreeRegressor(
    max_depth=3,
    random_state=42
)

# Step 4: Train the model
model.fit(X, y)

# Step 5: Extract feature importance
importance_values = model.feature_importances_

# Step 6: Create a feature importance DataFrame
importance_df = pd.DataFrame({
    "Feature": X.columns,
    "Importance": importance_values
})

# Step 7: Sort features by importance
importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

print(importance_df)

# Step 8: Visualize feature importance
plt.figure(figsize=(9, 5))

plt.barh(
    importance_df["Feature"],
    importance_df["Importance"]
)

plt.xlabel("Feature Importance")
plt.ylabel("Features")
plt.title("Decision Tree Feature Importance")

plt.gca().invert_yaxis()
plt.show()
```

### Step 1: Understand the Concept

**Random Forest Feature Importance**

Random Forest Feature Importance is a technique used to measure how much each input feature contributes to the predictions of a trained Random Forest model.

Random Forest is an ensemble learning algorithm that combines multiple Decision Trees to produce predictions.

Each tree learns from different samples of the training data and may use different features for splitting.

**How Random Forest Feature Importance Works:**

1. Random Forest builds multiple Decision Trees.
2. Each tree uses features to split the training data.
3. The model calculates impurity reduction for each feature in each tree.
4. The importance scores are combined across all trees.
5. The final scores are normalized.

**Important Points:**

- Features that contribute more to impurity reduction generally receive higher importance.
- Features that are not used for splitting may receive zero importance.
- Importance is aggregated across multiple trees.
- Random Forest can capture non-linear relationships and feature interactions.

**Types of Random Forest Feature Importance:**

1. Mean Decrease in Impurity (MDI):
   Calculates importance using impurity reduction across the trees.

2. Permutation Importance:
   Measures how much model performance decreases when a feature's values are shuffled.

**Note:**

This topic focuses on impurity-based Random Forest importance. Permutation Importance will be covered separately.

A high importance score does not prove that a feature causes changes in the target variable.

### Step 1: Understand the Concept

**Permutation Importance**

Permutation Importance is a model interpretation technique used to measure how much a feature contributes to a trained machine learning model's predictive performance.

It works by randomly shuffling the values of one feature and checking how much the model's performance changes.

**How Permutation Importance Works:**

1. Train a machine learning model using the dataset.
2. Calculate the model's original performance.
3. Select one feature.
4. Randomly shuffle the values of that feature.
5. Calculate the model's performance again.
6. Compare the original and shuffled performance.
7. Repeat the process for other features.

**Importance Calculation:**

For a scoring metric where higher values indicate better performance:

Permutation Importance = Original Score - Shuffled Score

**Interpretation:**

- Large positive importance: The feature is useful for the model's predictions.
- Importance close to zero: Shuffling the feature has little effect on performance.
- Negative importance: The model performs better after shuffling the feature.

**Important Points:**

- Permutation Importance can be used with different types of trained models.
- It measures the effect of a feature on model performance.
- It does not depend on the internal structure of a specific algorithm.
- Results can vary because feature values are randomly shuffled.

**Note:**

Permutation Importance measures predictive usefulness, not causation.



### Step 2: Demonstrate the Concept

**Real-World Example: Customer Spending Prediction**

An online shopping company builds a machine learning model to predict customer spending.

The model uses:

|---|---|

**Before Shuffling:**

**After Shuffling Purchase_Frequency:**

- Shuffled R² Score = 0.55

```python

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.inspection import permutation_importance
from sklearn.metrics import r2_score

# Step 1: Create sample customer data
data = {
    "Customer_Age": [22, 25, 28, 32, 35, 38, 42, 45, 24, 30, 34, 40,
                     26, 31, 36, 43, 29, 37, 41, 33],
    "Annual_Income": [20000, 25000, 30000, 40000, 45000, 55000,
                      65000, 75000, 22000, 35000, 48000, 70000,
                      28000, 38000, 52000, 68000, 32000, 58000,
                      72000, 42000],
    "Purchase_Frequency": [2, 3, 4, 5, 6, 7, 8, 9, 2, 4,
                           6, 8, 3, 5, 7, 9, 4, 7, 8, 6],
    "Customer_Visits": [3, 4, 5, 6, 7, 8, 9, 10, 4, 5,
                        7, 9, 4, 6, 8, 10, 5, 8, 9, 7],
    "Customer_Spending": [3000, 4000, 5000, 7000, 9000, 11000,
                          14000, 17000, 3500, 6000, 10000, 16000,
                          4500, 7500, 12000, 15500, 5500, 12500,
                          16500, 8500]
}

df = pd.DataFrame(data)

# Step 2: Separate input features and target
X = df.drop("Customer_Spending", axis=1)
y = df["Customer_Spending"]

# Step 3: Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Step 4: Create a Random Forest model
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

# Step 5: Train the model
model.fit(X_train, y_train)

# Step 6: Calculate original model performance
original_score = r2_score(
    y_test,
    model.predict(X_test)
)

print("Original R2 Score:", original_score)

# Step 7: Calculate permutation importance
result = permutation_importance(
    model,
    X_test,
    y_test,
    n_repeats=10,
    random_state=42,
    scoring="r2"
)

# Step 8: Create importance DataFrame
importance_df = pd.DataFrame({
    "Feature": X.columns,
    "Importance": result.importances_mean,
    "Standard_Deviation": result.importances_std
})

# Step 9: Sort features by importance
importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

print(importance_df)

# Step 10: Visualize permutation importance
plt.figure(figsize=(9, 5))

plt.barh(
    importance_df["Feature"],
    importance_df["Importance"],
    xerr=importance_df["Standard_Deviation"]
)

plt.xlabel("Permutation Importance (R2 Score Decrease)")
plt.ylabel("Features")
plt.title("Permutation Feature Importance")

plt.gca().invert_yaxis()
plt.show()
```

### Step 1: Understand the Concept

**Feature Importance vs Feature Selection**

Feature Importance and Feature Selection are two different techniques used to understand and improve the input features of a Machine Learning model.

**1. Feature Importance**

Feature Importance measures how much each input feature contributes to a trained model's predictions.

Purpose:
- Understand the contribution of individual features.
- Identify which features are useful for predictions.
- Interpret model behavior.
- Compare the relative importance of features.

Examples:
- Decision Tree Feature Importance.
- Random Forest Feature Importance.
- Permutation Importance.
- Coefficient-Based Importance.

**2. Feature Selection**

Feature Selection is the process of selecting a useful subset of input features and removing unnecessary or less useful features.

Purpose:
- Reduce the number of input features.
- Remove irrelevant or redundant features.
- Reduce model complexity.
- Improve training efficiency.
- Help reduce overfitting in some situations.

**Common Feature Selection Methods:**

- Filter Methods: Select features using statistical measures.
- Wrapper Methods: Evaluate different feature subsets using model performance.
- Embedded Methods: Perform feature selection during model training.

**Relationship Between Them:**

1. Train a model and calculate feature importance.
2. Identify features with potentially low predictive value.
3. Evaluate the features using validation data.
4. Select a suitable subset of features.
5. Train and evaluate the model again.

**Important Note:**

Feature Importance measures contribution, while Feature Selection decides which features to retain for model training.

A feature with low importance should not automatically be removed without validation.

Interpreting Feature Importance means understanding what the importance scores tell us about the input features used by a trained Machine Learning model.

Feature importance scores help us identify which features contribute more or less to a model's predictions.

**How to Interpret Feature Importance:**

1. Identify the feature with the highest importance score.
2. Compare the importance scores of all features.
3. Understand which features contribute more to the model.
4. Check whether the importance scores are reliable.
5. Consider business meaning and model performance.

**Understanding Importance Scores:**

- High Importance: The feature contributes more according to the selected importance method.
- Low Importance: The feature contributes less according to the selected importance method.
- Zero Importance: The selected method does not assign measurable importance to the feature.
- Negative Permutation Importance: Shuffling the feature improves the measured model performance.

**Important Interpretation Rules:**

1. Importance scores depend on the model and calculation method.
2. Importance scores do not always indicate whether a feature increases or decreases the prediction.
3. Correlated features can affect importance rankings.
4. Low importance does not always mean a feature is useless.
5. Importance scores do not prove causation.

**Feature Importance vs Causation:**

Feature Importance explains how useful a feature is for a model's predictions.

Causation explains whether changing one variable directly produces a change in another variable.

For example, Annual_Income may have high importance in predicting Customer_Spending. However, this does not prove that increasing annual income directly causes customer spending to increase.

Other factors may influence both variables.

Causal relationships require appropriate causal analysis or controlled experiments.

### Step 1: Understand the Concept

**Limitations of Feature Importance**

Feature Importance helps identify how much input features contribute to a Machine Learning model's predictions.

However, feature importance methods have certain limitations that can affect how we interpret their results.

**Common Limitations:**

1. **Model Dependency**
   - Different algorithms can produce different importance scores.
   - A feature may be important for one model but less important for another.

2. **Correlation Between Features**
   - Highly correlated features may share or divide importance.
   - One feature may appear less important because another correlated feature provides similar information.

3. **Bias Towards Certain Features**
   - Impurity-based tree importance can favor numerical features with many possible split points.
   - This can affect importance rankings.

4. **No Causal Relationship**
   - High importance does not prove that a feature causes changes in the target.
   - Feature importance identifies predictive patterns, not cause-and-effect relationships.

5. **Importance Does Not Show Direction**
   - A high importance score does not automatically tell whether a feature increases or decreases the prediction.
   - Additional interpretation is required.

6. **Dataset Dependency**
   - Importance scores depend on the dataset used for training and evaluation.
   - Different datasets may produce different importance rankings.

7. **Unstable Importance Scores**
   - Changes in training data, model settings, or random sampling may change importance scores.

8. **Low Importance Does Not Always Mean Useless**
   - Some features may provide useful information when combined with other features.
   - Removing them without validation may reduce model performance.

**How to Handle These Limitations:**

- Compare multiple importance methods.
- Check relationships between input features.
- Evaluate model performance using validation data.
- Understand the business meaning of each feature.
- Check for data leakage.
- Avoid making causal conclusions from importance scores.



### Step 2: Demonstrate the Concept

**Real-World Example: Customer Spending Prediction**

An online shopping company builds a Machine Learning model to predict customer spending.

The company uses:

| Feature | Description |
|---|---|
| Annual_Income | Customer's yearly income |
| Monthly_Income | Customer's monthly income |
| Purchase_Frequency | Number of purchases |
| Customer_Visits | Number of website visits |
| Customer_Age | Age of the customer |

**Problem 1: Correlated Features**

Annual_Income and Monthly_Income contain similar information.

The model may distribute importance between these two features.

As a result, one feature may receive a lower importance score even though both contain useful information.

**Problem 2: Model Dependency**

A Decision Tree and a Random Forest may produce different importance rankings.

This happens because each algorithm learns patterns differently.

**Problem 3: No Causal Evidence**

Suppose Annual_Income receives the highest importance score.

This does not prove that increasing annual income directly causes customer spending to increase.

Other factors, such as customer preferences and purchasing behavior, may influence spending.

**Problem 4: Dataset Dependency**

If the company trains the model using a different group of customers, the importance rankings may change.

**Business Understanding:**

Feature importance is useful for understanding model behavior, but it should not be the only factor used to make business decisions.

The company should also evaluate model performance, data quality, feature relationships, and real-world evidence.


```python

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor
from sklearn.inspection import permutation_importance
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

# Step 1: Create sample customer data
data = {
    "Customer_Age": [22, 25, 28, 32, 35, 38, 42, 45, 24, 30,
                     34, 40, 26, 31, 36, 43, 29, 37, 41, 33],
    "Annual_Income": [20000, 25000, 30000, 40000, 45000, 55000,
                      65000, 75000, 22000, 35000, 48000, 70000,
                      28000, 38000, 52000, 68000, 32000, 58000,
                      72000, 42000],
    "Monthly_Income": [20000/12, 25000/12, 30000/12, 40000/12,
                       45000/12, 55000/12, 65000/12, 75000/12,
                       22000/12, 35000/12, 48000/12, 70000/12,
                       28000/12, 38000/12, 52000/12, 68000/12,
                       32000/12, 58000/12, 72000/12, 42000/12],
    "Purchase_Frequency": [2, 3, 4, 5, 6, 7, 8, 9, 2, 4,
                           6, 8, 3, 5, 7, 9, 4, 7, 8, 6],
    "Customer_Visits": [3, 4, 5, 6, 7, 8, 9, 10, 4, 5,
                        7, 9, 4, 6, 8, 10, 5, 8, 9, 7],
    "Customer_Spending": [3000, 4000, 5000, 7000, 9000, 11000,
                          14000, 17000, 3500, 6000, 10000, 16000,
                          4500, 7500, 12000, 15500, 5500, 12500,
                          16500, 8500]
}

df = pd.DataFrame(data)

# Step 2: Separate input features and target
X = df.drop("Customer_Spending", axis=1)
y = df["Customer_Spending"]

# Step 3: Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# Step 4: Train a Random Forest model
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X_train, y_train)

# Step 5: Calculate model performance
predictions = model.predict(X_test)

score = r2_score(y_test, predictions)

print("Model R2 Score:", score)

# Step 6: Calculate impurity-based importance
impurity_df = pd.DataFrame({
    "Feature": X.columns,
    "Impurity_Importance": model.feature_importances_
})

# Step 7: Calculate permutation importance
permutation_result = permutation_importance(
    model,
    X_test,
    y_test,
    n_repeats=20,
    random_state=42,
    scoring="r2"
)

permutation_df = pd.DataFrame({
    "Feature": X.columns,
    "Permutation_Importance": permutation_result.importances_mean
})

# Step 8: Compare both importance methods
comparison_df = impurity_df.merge(
    permutation_df,
    on="Feature"
)

comparison_df = comparison_df.sort_values(
    by="Permutation_Importance",
    ascending=False
)

print("\nImportance Comparison:")
print(comparison_df)

# Step 9: Visualize importance comparison
comparison_df.plot(
    x="Feature",
    y=["Impurity_Importance", "Permutation_Importance"],
    kind="bar",
    figsize=(10, 5)
)

plt.xlabel("Features")
plt.ylabel("Importance Score")
plt.title("Comparison of Feature Importance Methods")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
