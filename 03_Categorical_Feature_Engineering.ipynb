# Topic 1 – Categorical Variables

## Step 1 – Understand the Concept

Categorical Variables are data columns that represent groups, labels, or categories instead of numerical measurements.

They describe the type or classification of an observation.

Examples:
- Gender: Male, Female
- Country: India, USA, UK
- PaymentMethod: Cash, Credit Card, UPI
- ProductCategory: Electronics, Clothing, Furniture

Categorical variables are commonly used in business applications and machine learning datasets.

Types of Categorical Variables:

1. Nominal Variables:
   Categories without any natural order.

2. Ordinal Variables:
   Categories with a meaningful order.

3. Binary Variables:
   Variables containing only two possible categories.

Why Are Categorical Variables Important?

Machine learning algorithms generally require numerical input. Therefore, categorical values often need to be converted into numerical representations using suitable encoding techniques.

The encoding method should be selected based on the meaning and structure of the categories.

## Step 2 – Demonstrate the Concept

Simple Example:

- UPI
- Cash
- Credit Card

These values represent different payment categories rather than numerical quantities.

Real-World Example:

An online shopping platform stores customer information such as:
- Country
- PaymentMethod
- ProductCategory
- DeliveryType

These columns help describe customer transactions.

Business Example:

A retail company can analyze product categories to understand which types of products customers purchase most frequently.

AI/ML Use Case:

A machine learning model predicting customer purchase behavior may use PaymentMethod, ProductCategory, and Country as input features.

Why Is Encoding Required?

A machine learning algorithm cannot directly perform mathematical calculations on text categories such as UPI or Cash.

Encoding converts these categories into numerical representations that algorithms can process.

When to Use:

Categorical variables should be identified whenever a dataset contains labels or group-based information that may help explain or predict the target.

Important Consideration:

Not every categorical variable should be encoded in the same way. The correct technique depends on whether the categories have an order, how many unique categories exist, and the information available during prediction.

# Topic 2 – Nominal Variables

## Step 1 – Understand the Concept

Nominal Variables are categorical variables that contain categories without any natural order or ranking.

Each category represents a different group or label.

Examples:

- PaymentMethod: UPI, Cash, Credit Card
- Country: India, USA, UK
- ProductCategory: Electronics, Clothing, Furniture
- DeliveryType: Standard, Express

Important Characteristics:

- Categories do not have a meaningful order.
- One category is not greater or smaller than another.
- Categories represent different groups.
- Mathematical comparisons between category labels are not meaningful.

Common Encoding Techniques:

1. One-Hot Encoding:
   Creates separate binary columns for categories.

2. Frequency Encoding:
   Replaces each category with its frequency in the dataset.

3. Target Encoding:
   Replaces each category with a statistic calculated from the target variable.

When to Use:

Nominal variables should be identified when categories represent names, types, or groups without any meaningful ranking.

The encoding technique should be selected based on the number of categories, dataset size, and prediction requirements.

## Step 2 – Demonstrate the Concept

Simple Example:

Consider the following payment methods:

Credit Card

These categories do not have a natural order.

UPI is not greater than Cash, and Cash is not greater than Credit Card.

Real-World Example:

An e-commerce company may store the country from which each customer places an order.

Countries such as India, USA, and UK are different categories without a natural ranking.

Business Example:

A retail business can analyze product categories to understand sales performance across Electronics, Clothing, and Furniture.

AI/ML Use Case:

A machine learning model predicting customer purchasing behavior may use PaymentMethod or ProductCategory as input features.

One-Hot Encoding Example:

PaymentMethod:

UPI        → [1, 0, 0]
Cash       → [0, 1, 0]
Credit Card → [0, 0, 1]

Each category receives its own binary representation.

When to Use One-Hot Encoding:

- When the number of unique categories is relatively small.
- When categories do not have a meaningful order.
- When the model should not interpret category codes as rankings.

Why to Use:

One-Hot Encoding represents categories without creating an artificial numerical order.

Important Consideration:

When a nominal column contains many unique categories, One-Hot Encoding can create too many columns. Other techniques may be more suitable depending on the dataset and model.

```python
import pandas as pd

# Create a sample e-commerce dataset
df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105],
    "PaymentMethod": [
        "UPI",
        "Cash",
        "Credit Card",
        "UPI",
        "Cash"
    ]
})

# Apply One-Hot Encoding
encoded_df = pd.get_dummies(
    df,
    columns=["PaymentMethod"],
    dtype=int
)

# Display the encoded dataset
print(encoded_df)
```

# Topic 3 – Ordinal Variables

## Step 1 – Understand the Concept

Ordinal Variables are categorical variables whose categories have a meaningful order or ranking.

The categories can be arranged from lower to higher or from one level to another.

Examples:

- SatisfactionLevel: Low, Medium, High
- ProductQuality: Poor, Average, Good, Excellent
- EducationLevel: School, Undergraduate, Postgraduate
- CustomerRating: 1 Star, 2 Stars, 3 Stars, 4 Stars, 5 Stars

Important Characteristics:

- Categories have a meaningful order.
- Categories can be ranked.
- The distance between categories is not necessarily equal.
- Numerical values should represent the correct category order.

Common Encoding Technique:

Ordinal Encoding assigns numerical values based on the natural order of categories.

Example:

Low = 1
Medium = 2
High = 3

When to Use:

Ordinal variables should be identified when categories have a meaningful ranking.

Why to Use Ordinal Encoding:

It converts ordered categories into numerical values while preserving their ranking information.

Important Consideration:

Do not use arbitrary numerical values for categories without a meaningful order. Also, do not assume that the distance between ordinal categories is equal.

## Step 2 – Demonstrate the Concept

Simple Example:

Consider three customer satisfaction levels:

Low
Medium
High

Their natural order is:

Low < Medium < High

Using Ordinal Encoding:

Low = 1
Medium = 2
High = 3

Real-World Example:

An online shopping platform may collect customer feedback using satisfaction levels such as Poor, Average, Good, and Excellent.

Business Example:

A retail company can analyze customer satisfaction levels to understand service quality and identify areas that need improvement.

AI/ML Use Case:

A machine learning model predicting customer satisfaction may use ordered feedback categories as numerical input features.

When to Use Ordinal Encoding:

- When categories have a meaningful order.
- When the ranking information should be preserved.
- When converting ordered categorical data into numerical values.

Difference Between Nominal and Ordinal Variables:

Nominal:
- Categories have no natural order.
- Example: UPI, Cash, Credit Card.
- One-Hot Encoding is commonly suitable.

Ordinal:
- Categories have a meaningful order.
- Example: Low, Medium, High.
- Ordinal Encoding is commonly suitable.

Why Is This Difference Important?

If nominal categories are assigned arbitrary numbers, a model may interpret them as having an artificial order.

For ordinal categories, numerical values can preserve the meaningful ranking.

Important Consideration:

Ordinal Encoding does not prove that the difference between Low and Medium is equal to the difference between Medium and High.

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# Create a sample e-commerce dataset
df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105],
    "SatisfactionLevel": [
        "Medium",
        "High",
        "Low",
        "Excellent",
        "Good"
    ]
})

# Define the natural category order
category_order = [
    "Low",
    "Medium",
    "Good",
    "High",
    "Excellent"
]

# Create the ordinal encoder
encoder = OrdinalEncoder(
    categories=[category_order]
)

# Apply ordinal encoding
df["SatisfactionEncoded"] = encoder.fit_transform(
    df[["SatisfactionLevel"]]
).astype(int)

# Display the updated dataset
print(df)
```

# Topic 4 – Binary Variables

## Step 1 – Understand the Concept

Binary Variables are categorical variables that contain only two possible categories or values.

They represent two different states or conditions.

Examples:

- Purchased: Yes, No
- Membership: Active, Inactive
- PaymentCompleted: True, False
- CustomerSubscribed: 1, 0

Important Characteristics:

- A binary variable contains only two categories.
- It represents two possible conditions.
- It can be converted into numerical values using binary encoding.
- It is commonly represented using 0 and 1.

Common Encoding Method:

Binary Encoding for two categories:

Yes = 1
No = 0

When to Use:

Binary variables should be identified when a feature contains exactly two meaningful categories.

Why to Use:

Converting binary categories into 0 and 1 makes them suitable for machine learning algorithms that require numerical inputs.

Important Consideration:

The meaning of 0 and 1 should be clearly defined. A binary variable may also be stored as a Boolean or numerical column, so data type alone is not enough to identify it.

## Step 2 – Demonstrate the Concept

Simple Example:

Consider a customer subscription feature:

Subscribed:
- Yes
- No

Binary Encoding:

Yes = 1
No = 0

Real-World Example:

An e-commerce platform may store whether a customer has completed a purchase.

PurchaseCompleted:
- Yes
- No

Business Example:

A retail company can analyze whether customers have subscribed to its membership program to understand membership participation.

AI/ML Use Case:

A machine learning model predicting customer retention may use a historical membership status as an input feature.

When to Use:

- When a categorical variable contains only two possible categories.
- When representing Yes/No or True/False information.
- When a simple numerical representation is sufficient.

Why to Use:

It converts two categories into a compact numerical feature without creating multiple columns.

Difference Between Binary and One-Hot Encoding:

Binary Encoding for two categories:
- Usually creates one column containing 0 and 1.
- Example: Yes = 1, No = 0.

One-Hot Encoding:
- Creates a separate binary column for each category.
- With two categories, it may create two columns.
- One column can be sufficient when the categories are mutually exclusive and exhaustive.

Important Consideration:

Choose the positive and negative category meanings consistently. Do not encode categories in a way that reverses their meaning.

```python
import pandas as pd

# Create a sample e-commerce dataset
df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105],
    "MembershipStatus": [
        "Yes",
        "No",
        "Yes",
        "No",
        "Yes"
    ]
})

# Create a mapping for binary categories
membership_mapping = {
    "No": 0,
    "Yes": 1
}

# Apply binary encoding
df["MembershipEncoded"] = (
    df["MembershipStatus"].map(membership_mapping)
)

# Display the updated dataset
print(df)
```

# Topic 5 – One-Hot Encoding

## Step 1 – Understand the Concept

One-Hot Encoding is a categorical encoding technique that converts each unique category into a separate binary column.

Each category is represented using:
- 1 if the observation belongs to that category.
- 0 if the observation does not belong to that category.

Example:

PaymentMethod:

UPI
Cash
Credit Card

After One-Hot Encoding:

| PaymentMethod | Cash | Credit Card | UPI |
|---|---:|---:|---:|
| UPI | 0 | 0 | 1 |
| Cash | 1 | 0 | 0 |
| Credit Card | 0 | 1 | 0 |

Important Characteristics:

- Each unique category creates a separate column.
- Encoded values are represented using 0 and 1.
- It does not create an artificial ranking between nominal categories.
- It is commonly used for nominal categorical variables.

When to Use:

- When a categorical column has a relatively small number of unique categories.
- When categories do not have a natural order.
- When a model requires numerical input.

Why to Use:

It converts text categories into numerical features while preserving the identity of each category.

Important Consideration:

One-Hot Encoding can create many columns when a feature has high cardinality. In such cases, other encoding techniques may be more suitable.

## Step 2 – Demonstrate the Concept

Simple Example:

Consider a column called Color:

Red
Blue
Green

After One-Hot Encoding:

| Color | Blue | Green | Red |
|---|---:|---:|---:|
| Red | 0 | 0 | 1 |
| Blue | 1 | 0 | 0 |
| Green | 0 | 1 | 0 |

Real-World Example:

An e-commerce platform stores the payment methods selected by customers, such as UPI, Cash, and Credit Card.

One-Hot Encoding converts these payment methods into separate numerical columns.

Business Example:

A retail business can analyze sales across different product categories without assigning an artificial numerical ranking to product types.

AI/ML Use Case:

A customer purchase prediction model can use encoded payment methods or product categories as input features.

When to Use:

- For nominal categorical variables.
- When the number of unique categories is manageable.
- When each category needs an independent numerical representation.

Why to Use:

It allows machine learning algorithms to process categorical information without treating category labels as ordered numerical values.

Difference from Ordinal Encoding:

One-Hot Encoding:
- Creates separate binary columns.
- Does not assign a ranking.
- Suitable for nominal categories.

Ordinal Encoding:
- Usually creates one numerical column.
- Assigns values based on a meaningful category order.
- Suitable for ordinal categories.

Important Considerations:

- The number of columns increases with the number of unique categories.
- One category may be dropped when appropriate for models with an intercept to avoid perfect multicollinearity.
- During model deployment, unseen categories must be handled consistently.

# Topic 6 – Ordinal Encoding

## Step 1 – Understand the Concept

Ordinal Encoding is a technique used to convert categorical values into numerical values while preserving their natural order.

It is suitable for categorical variables that have a meaningful ranking or sequence.

For example:
- Low = 0
- Medium = 1
- High = 2

The numerical values represent the order of the categories.

**Why is Ordinal Encoding used?**

- It converts ordered categories into numerical values.
- It preserves the ranking information.
- It helps Machine Learning algorithms work with ordered categorical data.

**When should it be used?**

Use Ordinal Encoding when categories have a meaningful order, such as:
- Customer satisfaction levels
- Education levels
- Product quality ratings
- Service priority levels

The category order must be defined using domain knowledge. The numerical differences do not necessarily represent equal differences in meaning.

## Step 2 – Demonstrate the Concept

**Simple Example:**

Consider three customer satisfaction levels:

- Low
- Medium
- High

Their natural order is:

Low < Medium < High

We can represent them numerically as:

- Low = 0
- Medium = 1
- High = 2

**Real-World Example:**

A customer feedback system collects satisfaction levels such as Low, Medium, and High. Ordinal Encoding converts these values into numerical representations while maintaining their order.

**Business Example:**

An e-commerce company can convert customer satisfaction levels into numerical values to analyze customer experience.

**AI/ML Use Case:**

A Machine Learning model can use encoded satisfaction levels as input features to identify patterns related to customer behavior or repeat purchases.

**Why is it useful?**

It preserves the meaningful order of categories without creating multiple columns for each category.

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# Create a sample dataset
df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105],
    "SatisfactionLevel": ["Low", "Medium", "High", "Medium", "Low"]
})

# Define the natural order of categories
category_order = [["Low", "Medium", "High"]]

# Create the OrdinalEncoder
encoder = OrdinalEncoder(categories=category_order)

# Convert categorical values into numerical values
df["SatisfactionEncoded"] = encoder.fit_transform(
    df[["SatisfactionLevel"]]
).astype(int)

# Display the result
print(df)
```

# Topic 8 – Frequency Encoding

Frequency Encoding converts these categories into numerical values:

**Why is Frequency Encoding used?**

- It converts categorical values into numerical representations.
- It captures how frequently each category appears.
- It can reduce the number of additional columns compared with One-Hot Encoding.

**When should it be used?**

It can be useful for categorical columns with many unique categories, such as:
- Product categories
- Customer locations
- Payment methods
- Product codes

**Important Note:**

Frequency values represent category occurrence counts, not category rankings or importance. The frequency calculation must use training data only in a Machine Learning workflow.

## Step 2 – Demonstrate the Concept

Consider a dataset containing these payment methods:

UPI, Cash, UPI, Credit Card, UPI, Cash, UPI, Credit Card, Cash

The frequency of each category is:

Each category is replaced with its frequency.

**Real-World Example:**

- Label Encoding assigns numerical identifiers to categories.

Frequency Encoding is useful when category frequency contains meaningful information, but it may cause different categories with the same frequency to receive the same value.

# Topic 9 – Target Encoding

## Step 1 – Understand the Concept

Target Encoding is a categorical encoding technique that replaces each category with the average value of the target variable for that category.

For example, consider a customer purchase prediction dataset:

| ProductCategory | RepeatPurchase |
|---|---:|
| Electronics | 1 |
| Electronics | 0 |
| Clothing | 1 |
| Clothing | 1 |
| Electronics | 1 |

The average repeat purchase rate is:

- Electronics = 2/3 = 0.67
- Clothing = 2/2 = 1.00

Target Encoding replaces each category with its corresponding target average.

**Why is Target Encoding used?**

- It captures the relationship between categorical features and the target.
- It converts categories into numerical values.
- It can represent high-cardinality categorical features without creating many columns.

**When should it be used?**

It can be useful for categorical columns with many unique categories, such as:
- Product categories
- Customer locations
- Product brands
- Store identifiers

**Important Note:**

Target Encoding uses target information. Therefore, it has a high risk of target leakage and overfitting if implemented incorrectly.

Use training data only, and prefer out-of-fold or cross-validation-based encoding for training features.

## Step 2 – Demonstrate the Concept

**Simple Example:**

Consider a dataset containing product categories and repeat purchase labels.

- Electronics has 3 customers, and 2 made repeat purchases.
- Clothing has 2 customers, and both made repeat purchases.

The average repeat purchase rate is:

- Electronics = 2 / 3 = 0.67
- Clothing = 2 / 2 = 1.00

These averages are used as encoded values.

**Real-World Example:**

An e-commerce company can calculate the historical purchase rate for each product category.

**Business Example:**

A retail business can understand which product categories have historically shown higher repeat purchase rates.

**AI/ML Use Case:**

A classification model can use category-level target averages as numerical input features to learn patterns related to repeat purchases.

**Why is it useful?**

It captures category-level target information in a single numerical column, which can be especially helpful when a categorical feature has many unique values.

**Difference from Frequency Encoding:**

- Frequency Encoding uses the number of category occurrences.
- Target Encoding uses the average target value for each category.

**Important ML Consideration:**

For training data, use out-of-fold target encoding or another leakage-safe method. Calculate validation and test encodings using only the corresponding training data. Never use the validation or test target values to calculate their encodings.

```python
import pandas as pd

# Create a sample dataset
df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105, 106],
    "ProductCategory": [
        "Electronics", "Electronics", "Clothing",
        "Clothing", "Electronics", "Furniture"
    ],
    "RepeatPurchase": [1, 0, 1, 1, 1, 0]
})

# Calculate the mean target value for each category
target_map = df.groupby("ProductCategory")[
    "RepeatPurchase"
].mean()

# Create a target-encoded feature
df["CategoryTargetEncoded"] = df["ProductCategory"].map(
    target_map
)

# Display the result
print(df)

# Display the target mapping
print("\nTarget Encoding Mapping:")
print(target_map)
```

- `RepeatPurchase`: Binary target variable where 1 represents repeat purchase and 0 represents no repeat purchase.

**3. Calculating the Target Average**

`df.groupby("ProductCategory")["RepeatPurchase"].mean()`

- `groupby()` groups the records based on product category.
- `["RepeatPurchase"]` selects the target column.
- `mean()` calculates the average target value for each category.

The result is stored in `target_map`.

**4. Creating the Target-Encoded Feature**

`df["ProductCategory"].map(target_map)`

The `map()` method replaces each product category with its corresponding average target value.

For example:
- Electronics = 0.67
- Clothing = 1.00
- Furniture = 0.00

**5. Creating a New Column**

`df["CategoryTargetEncoded"]` stores the numerical target-encoded values.

The original category column is retained for comparison.

**Expected Output:**

| ProductCategory | RepeatPurchase | CategoryTargetEncoded |
|---|---:|---:|
| Electronics | 1 | 0.67 |
| Electronics | 0 | 0.67 |

**Why was this feature created?**

The original product category contains text values. The new feature represents the historical average repeat purchase rate for each category.

**Important Production Workflow:**

This example demonstrates the basic calculation on a small dataset. It is not a leakage-safe training implementation because each training row contributes to the average used to encode its own category.

For a real Machine Learning workflow:

- Split the data into training and validation/test sets first.
- Generate training encodings using out-of-fold calculations.
- Calculate category mappings from the full training data for validation and test data.
- Use smoothing or regularization for categories with very few records.
- Handle categories not seen during training.

This prevents target leakage and reduces overfitting.

## Step 4 – Justify Your Feature

**Feature Name:** CategoryTargetEncoded

**Source Columns:** ProductCategory, RepeatPurchase

CategoryTargetEncoded = Mean of RepeatPurchase for each ProductCategory.

Formula:

Category Target Mean = Sum of RepeatPurchase Values / Number of Records in the Category

**Reason for Creation:**

To represent the historical relationship between product categories and repeat purchase behavior as a numerical feature.

**Expected Relationship with Target:**

Categories with higher historical repeat purchase rates may be associated with a greater probability of future repeat purchases. This relationship must be validated using unseen data.

**Potential Business Meaning:**

The feature helps businesses identify product categories that have historically attracted more repeat purchases.

**Potential ML Value:**

**Leakage Risk:**

Target Encoding has a high leakage risk because it directly uses target values.

If a row's own target contributes to its encoded value, the model may learn information that would not be available during prediction.

Use out-of-fold encoding for training data and training-only mappings for validation and test data.

**Limitations:**

Categories with very few records may produce unreliable target averages. Smoothing can help reduce the effect of small category groups.

**Final Decision:** Retain with Leakage-Safe Encoding

**Justification:**

Retain this feature only when it is generated using a leakage-safe encoding process and provides useful performance on unseen validation data.

**Why is Rare Category Grouping used?**

```python
import pandas as pd

# Create a sample dataset
df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105, 106, 107, 108, 109],
    "ProductCategory": [
        "Electronics", "Electronics", "Clothing",
        "Clothing", "Furniture", "Books",
        "Toys", "Electronics", "Clothing"
    ]
})

# Define the minimum frequency threshold
minimum_frequency = 2

# Calculate category frequencies
category_counts = df["ProductCategory"].value_counts()

# Identify categories that meet the threshold
common_categories = category_counts[
    category_counts >= minimum_frequency
].index

# Group rare categories into Other
df["GroupedCategory"] = df["ProductCategory"].where(
    df["ProductCategory"].isin(common_categories),
    "Other"
)

# Display the result
print(df)

# Display grouped category frequencies
print("\nGrouped Category Frequencies:")
print(df["GroupedCategory"].value_counts())
```

# Topic 11 – Category Cardinality

## Step 1 – Understand the Concept

Category Cardinality refers to the number of unique categories present in a categorical feature.

For example, consider a PaymentMethod column:

- UPI
- Cash
- Credit Card

This column has 3 unique categories. Therefore, its cardinality is 3.

**Types of Category Cardinality**

1. Low Cardinality:
   - Contains a small number of unique categories.
   - Example: Gender, PaymentMethod, MembershipStatus.

2. High Cardinality:
   - Contains a large number of unique categories.
   - Example: CustomerID, ProductCode, City, ProductName.

**Why is Category Cardinality important?**

- It helps understand categorical data complexity.
- It helps select an appropriate encoding technique.
- It helps identify features that may create too many encoded columns.
- It supports better feature engineering decisions.

**When should it be checked?**

Check cardinality before encoding categorical features, especially when using One-Hot Encoding.

**Important Note:**

High cardinality does not automatically mean a feature is useful or should be removed. Domain knowledge and prediction requirements must guide the decision.

## Step 2 – Demonstrate the Concept

**Simple Example:**

Consider two categorical columns:

PaymentMethod:
- UPI
- Cash
- Credit Card

City:
- Chennai
- Mumbai
- Delhi
- Bangalore
- Hyderabad
- Coimbatore

**Real-World Example:**

An e-commerce dataset may contain a PaymentMethod column with a few unique values and a ProductCode column with thousands of unique values.

**Business Example:**

A retail company can examine category cardinality to understand how many unique product types, brands, and locations are present.

**AI/ML Use Case:**

Before encoding categorical features, an AI/ML engineer checks cardinality to decide whether One-Hot Encoding, Frequency Encoding, Target Encoding, or Rare Category Grouping is more appropriate.

**Why is it useful?**

High-cardinality features can create a large number of columns when One-Hot Encoding is applied. This may increase memory usage and model complexity.

**Difference Between Low and High Cardinality:**

- Low Cardinality contains fewer unique categories and is often suitable for One-Hot Encoding.
- High Cardinality contains many unique categories and may require alternative encoding techniques.

The decision depends on the feature's meaning, dataset size, and model requirements.

```python
import pandas as pd

# Create a sample dataset
df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105, 106],
    "PaymentMethod": [
        "UPI", "Cash", "UPI", "Credit Card", "Cash", "UPI"
    ],
    "City": [
        "Chennai", "Mumbai", "Delhi",
        "Chennai", "Bangalore", "Hyderabad"
    ],
    "ProductCode": [
        "P101", "P102", "P103", "P104", "P105", "P106"
    ]
})

# Select categorical columns
categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

# Calculate unique category counts
cardinality = df[categorical_columns].nunique()

# Create a cardinality summary
cardinality_summary = pd.DataFrame({
    "UniqueCategories": cardinality,
    "CardinalityType": cardinality.apply(
        lambda count: "Low Cardinality" if count <= 3
        else "High Cardinality"
    )
})

# Display the result
print(cardinality_summary)
```

# Topic 12 – Handling Unknown Categories

## Step 1 – Understand the Concept

Unknown Categories are categorical values that appear in validation, testing, or new prediction data but were not present when the encoder was fitted on the training data.

For example:

Training Data:
- UPI
- Cash
- Credit Card

New Data:
- UPI
- Cash
- Wallet

Here, Wallet is an unknown category because it was not present in the training data.

**Why is Handling Unknown Categories important?**

- It prevents encoding errors when new categories appear.
- It supports real-world prediction scenarios.
- It helps maintain a consistent feature structure.
- It makes Machine Learning pipelines more reliable.

**Common Handling Methods:**

1. Ignore Unknown Categories:
   Use OneHotEncoder with `handle_unknown="ignore"`.

2. Group Unknown Categories:
   Replace unseen values with a predefined category such as Other, if the category mapping was designed during training.

3. Use an Unknown Category:
   Define an explicit unknown category during preprocessing.

**When should it be used?**

It is important when categorical features may contain new values during validation, testing, or real-world predictions.

**Important Note:**

Fit the encoder using training data only. Never fit a separate encoder on test data because this can produce inconsistent feature columns.

## Step 2 – Demonstrate the Concept

**Simple Example:**

Consider a payment method column.

Training categories:
- UPI
- Cash
- Credit Card

New customer payment method:
- Wallet

The Wallet category was not available during training.

If the encoder cannot handle unknown categories, the transformation may produce an error.

Using `OneHotEncoder(handle_unknown="ignore")` allows the transformation to continue.

**Real-World Example:**

An e-commerce platform initially supports UPI, Cash, and Credit Card. Later, it introduces Wallet as a new payment method.

The existing Machine Learning model may receive Wallet as an unseen category.

**Business Example:**

A retail company may receive new product categories or customer locations that were not available in its historical training data.

**AI/ML Use Case:**

A prediction pipeline must handle unseen categorical values without changing the expected feature structure.

**Why is it useful?**

It helps a trained model process new records even when some categorical values were not present during training.

**Important ML Consideration:**

Ignoring an unknown category prevents an error, but it does not teach the model the meaning of that category. The model receives an all-zero encoding for that feature's one-hot columns.

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Create training data
train_df = pd.DataFrame({
    "PaymentMethod": [
        "UPI", "Cash", "Credit Card", "UPI", "Cash"
    ]
})

# Create new data containing an unknown category
test_df = pd.DataFrame({
    "PaymentMethod": [
        "UPI", "Wallet", "Cash"
    ]
})

# Create the encoder
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

# Fit the encoder using training data only
train_encoded = encoder.fit_transform(
    train_df[["PaymentMethod"]]
)

# Transform new data using the same encoder
test_encoded = encoder.transform(
    test_df[["PaymentMethod"]]
)

# Get feature names
feature_names = encoder.get_feature_names_out(
    ["PaymentMethod"]
)

# Create encoded DataFrames
train_result = pd.DataFrame(
    train_encoded,
    columns=feature_names
)

test_result = pd.DataFrame(
    test_encoded,
    columns=feature_names
)

# Display results
print("Training Data Encoding:")
print(train_result)

print("\nNew Data Encoding:")
print(test_result)
```

## Step 3 – Implement the Concept

**1. Importing Libraries**

- `pandas` is used to create and manage the datasets.
- `OneHotEncoder` is imported from Scikit-learn to convert categorical values into binary columns.

**2. Creating Training Data**

`train_df` contains the payment methods available during model training.

The known categories are:
- UPI
- Cash
- Credit Card

**3. Creating New Data**

`test_df` contains payment methods from new records.

It includes Wallet, which was not available in the training data.

**4. Creating the Encoder**

`OneHotEncoder(handle_unknown="ignore", sparse_output=False)`

- `handle_unknown="ignore"` tells the encoder to ignore categories that were not present during fitting.
- `sparse_output=False` returns a dense numerical array instead of a sparse matrix.

**5. Fitting the Encoder**

`encoder.fit_transform(train_df[["PaymentMethod"]])`

The encoder learns the categories from the training data and converts them into binary columns.

**6. Transforming New Data**

`encoder.transform(test_df[["PaymentMethod"]])`

The same fitted encoder transforms the new data.

The unknown Wallet category does not produce an error. Its encoded values are all zeros across the learned payment-method columns.

**7. Getting Feature Names**

`encoder.get_feature_names_out(["PaymentMethod"])`

This returns the names of the generated encoded columns.

**8. Creating Encoded DataFrames**

`pd.DataFrame()` converts the encoded arrays into DataFrames with readable column names.

**Expected Output:**

Training Data Encoding:

| PaymentMethod_Cash | PaymentMethod_Credit Card | PaymentMethod_UPI |
|---:|---:|---:|
| 0 | 0 | 1 |
| 1 | 0 | 0 |
| 0 | 1 | 0 |
| 0 | 0 | 1 |
| 1 | 0 | 0 |

New Data Encoding:

| PaymentMethod_Cash | PaymentMethod_Credit Card | PaymentMethod_UPI |
|---:|---:|---:|
| 0 | 0 | 1 |

**Why was this feature created?**

It ensures that unseen payment methods can be transformed without changing the feature structure learned during training.

**ML Consideration:**

- Always fit the encoder using training data only.
- Use the same fitted encoder for validation, testing, and new prediction data.
- Monitor unknown categories because frequent unseen values may indicate changes in customer behavior or business operations.
- Consider a predefined Other category when unknown values need a meaningful representation.

## Step 4 – Justify Your Feature

**Feature Name:** Encoded PaymentMethod Columns

**Source Columns:** PaymentMethod

**Feature Logic / Formula:**

One-Hot Encoding creates one binary column for each category learned from training data.

If a category is unknown during transformation and `handle_unknown="ignore"` is used, all encoded columns for that feature receive zero.

**Reason for Creation:**

To handle new categorical values without causing encoding errors during prediction.

**Expected Relationship with Target:**

Known payment methods may have different relationships with customer purchasing behavior. Unknown categories require additional monitoring and cannot automatically be assumed to have the same relationship.

**Potential Business Meaning:**

The feature allows an e-commerce platform to process customer records even when new payment methods are introduced.

**Potential ML Value:**

It maintains consistent input columns between training and prediction and supports reliable transformation of unseen records.

**Leakage Risk:**

Fitting the encoder on the complete dataset before splitting can expose information from validation or test data.

Fit the encoder using training data only.

**Limitations:**

An unknown category encoded as all zeros does not provide its actual business meaning to the model. The model cannot learn a separate effect for an unseen category without suitable training examples.

**Final Decision:** Retain

**Justification:**

Retain unknown-category handling in the preprocessing pipeline to prevent transformation errors and maintain consistent feature columns during real-world predictions.

# Topic 13 – Combining Categories

Categories can be combined based on business knowledge, similarity, or their shared purpose.

- Credit Card
- Cash

**Why is Combining Categories used?**

- It simplifies categorical data.
- It combines related categories into meaningful groups.
- It reduces the number of unique categories.

**When should it be used?**

It can be useful when categories share a common business meaning, such as:

- Debit Card and Credit Card → Card Payment
- Mobile and Desktop → Digital Device
- North Chennai and South Chennai → Chennai Region
- Small and Medium → Lower Size Group, when business rules support this grouping.

```python
import pandas as pd

# Create a sample dataset
df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105, 106, 107],
    "PaymentMethod": [
        "UPI", "Debit Card", "Credit Card",
        "Cash", "Debit Card", "UPI", "Credit Card"
    ]
})

# Define business-based category mapping
category_mapping = {
    "Debit Card": "Card Payment",
    "Credit Card": "Card Payment",
    "UPI": "UPI",
    "Cash": "Cash"
}

# Create a new combined category feature
df["CombinedPaymentCategory"] = (
    df["PaymentMethod"].map(category_mapping)
)

# Display the result
print(df)

# Display combined category counts
print("\nCombined Category Counts:")
print(df["CombinedPaymentCategory"].value_counts())
```
