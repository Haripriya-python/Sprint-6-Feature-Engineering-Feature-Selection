# Topic 1 – What are Interaction Features?

## Step 1 – Understand the Concept

Interaction Features are new features created by combining two or more existing features using mathematical operations.

They help machine learning models understand how two or more variables work together to influence a target variable.

Why do we use Interaction Features?
- To identify relationships between multiple features.
- To capture combined effects that individual features may not represent.
- To improve the predictive information available to a model.
- To represent meaningful business relationships.

When do we use Interaction Features?
- When two features may influence an outcome together.
- When the relationship between features and the target is not fully represented by individual features.
- When domain knowledge suggests a meaningful combination.
- When model performance can benefit from additional combined information.

Simple Example:

Consider two features:
- Quantity = 5
- Price = 100

Interaction Feature:

TotalAmount = Quantity × Price

TotalAmount = 5 × 100 = 500

Real-World Example:

An online shopping platform combines product quantity and unit price to calculate the total amount of a purchase.

Business Example:

A company combines employee experience and salary information to study compensation patterns.

AI/ML Use Case:

Interaction features can help machine learning models capture combined effects in customer spending prediction, sales forecasting, and risk analysis.

When Are Interaction Features Useful?

- When a meaningful relationship exists between two features.
- When domain knowledge supports combining the features.
- When the new feature improves model performance on unseen data.
- When the interaction represents useful information that individual features may not capture.

When Can They Create Unnecessary Complexity?

- When unrelated features are combined without a clear reason.
- When too many combinations create a large number of features.
- When new features contain duplicate or redundant information.
- When features introduce noise or overfitting.
- When the interaction uses information unavailable at prediction time.

Important Note:

An interaction feature is not automatically useful just because it combines two variables.

Its usefulness must be evaluated using domain knowledge, validation performance, and leakage checks.

# Topic 3 – Feature + Feature

## Step 1 – Understand the Concept

Feature + Feature is an interaction feature technique in which two numerical features are added together to create a new feature.

The addition operation is useful when both features represent compatible quantities and their combined value has a meaningful business interpretation.

Why do we use Feature + Feature?

- To calculate a combined total from two related numerical features.
- To represent the overall value of multiple components.
- To create a meaningful feature that may help an ML model understand combined information.

When is it useful?

- When two features measure compatible quantities.
- When their sum represents a meaningful business value.
- When the combined value is available at the time of prediction.

Simple Example:

Savings + Monthly Deposit = Total Available Savings

Real-World Example:

Product Amount + Shipping Fee = Total Customer Cost

Business Example:

A company can combine product amount and shipping fee to understand the total amount charged to a customer.

AI/ML Use Case:

A machine learning model can use a meaningful combined cost feature to understand customer spending patterns or predict a related business outcome.

Important Considerations:

- Both features should have compatible units and meanings.
- Avoid adding unrelated features.
- Check whether an equivalent total feature already exists.
- Do not use information that becomes available only after the prediction time.
- Evaluate whether the new feature improves model performance.

## Step 2 – Demonstrate the Concept

Consider an e-commerce business that records the product amount and shipping fee for each order.

The business wants to calculate the total cost paid by the customer.

Example:

|---|---:|---:|---:|
| C001 | 500 | 50 | 550 |
| C002 | 1000 | 100 | 1100 |

Feature Creation:

Simple Explanation:

If a product costs ₹500 and its shipping fee is ₹50, the total customer cost is ₹550.

Real-World Usage:

Online shopping platforms calculate the final amount by combining the product cost and applicable delivery charges.

Business Usage:

The combined value helps businesses understand the total amount charged for an order.

AI/ML Usage:

The combined feature may help a model analyze customer spending or predict a related outcome, provided the information is available at prediction time.

```python
import pandas as pd

data = {
    "Customer": ["C001", "C002", "C003", "C004"],
    "ProductAmount": [500, 1000, 750, 1200],
    "ShippingFee": [50, 100, 50, 0]
}

df = pd.DataFrame(data)

df["TotalOrderCost"] = (
    df["ProductAmount"] + df["ShippingFee"]
)

print(df)
```

   - The DataFrame makes it easier to perform feature calculations.

4. Create the new feature:
   - df["TotalOrderCost"] creates a new column.
   - df["ProductAmount"] retrieves the product amount.
   - df["ShippingFee"] retrieves the shipping fee.
   - The + operator adds the two numerical columns row by row.

5. Display the result:
   - print(df) displays the original features and the newly created feature.

Why is addition used?

Addition is appropriate because the product amount and shipping fee are both monetary values. Their sum represents the total order cost.

Before Feature Creation:
The product amount and shipping fee are stored separately.

After Feature Creation:
A new TotalOrderCost feature represents their combined value.

AI/ML Application:

The combined feature can provide useful information about customer spending. However, its usefulness depends on the prediction objective and whether both source values are available when the prediction is made.

Important:
If TotalOrderCost itself is the prediction target, calculating it directly from ProductAmount and ShippingFee is not a meaningful independent prediction task. Also, do not use values that become available only after the target event.

## Step 4 – Justify Your Feature

TotalOrderCost

### Source Columns
- ProductAmount
- ShippingFee

### Reason for Creation
The feature combines the product amount and shipping fee into one value that represents the total cost charged to the customer.

### Expected Relationship with Target
If the target is a related customer spending or purchasing outcome, TotalOrderCost may provide useful information.

### Potential ML Value

### Leakage Risk
There is a leakage risk if the product amount or shipping fee is calculated or finalized only after the event being predicted.

The feature should only use information available at the prediction time.

### Final Decision (Retain/Remove)
Retain if both source features are meaningful, available at prediction time, and the combined feature provides useful information.

Remove if an equivalent total feature already exists or the new feature does not improve model performance.

# Topic 4 – Feature − Feature

## Step 1 – Understand the Concept

Feature − Feature is a feature engineering technique in which one numerical feature is subtracted from another to create a new feature.

The subtraction operation helps identify the difference or gap between two related numerical values.

Why do we use Feature − Feature?

- To calculate the difference between two related values.
- To identify increases or decreases.
- To understand gaps between actual and expected values.
- To create features that represent changes or variations.

When is it useful?

- When both features represent comparable quantities.
- When the difference has a meaningful interpretation.
- When the difference can provide useful information for the prediction task.

Simple Example:

Current Balance − Previous Balance = Balance Change

Real-World Example:

Actual Delivery Days − Expected Delivery Days = Delivery Delay

Business Example:

A company can compare actual delivery time with expected delivery time to identify delayed orders.

AI/ML Use Case:

A machine learning model can use delivery delay information to identify patterns associated with late deliveries.

Important Considerations:

- Both features should use compatible units.
- The subtraction order must match the intended meaning.
- Check whether the difference can be negative.
- Avoid subtracting unrelated features.
- Use only information available at prediction time.

## Step 2 – Demonstrate the Concept

Consider an e-commerce company that tracks the expected and actual delivery days for each order.

The company wants to calculate how many days early or late an order was delivered.

Example:

| Order | Actual Delivery Days | Expected Delivery Days | Delivery Difference |
|---|---:|---:|---:|
| O001 | 5 | 3 | 2 |
| O002 | 2 | 4 | -2 |
| O003 | 4 | 4 | 0 |
| O004 | 7 | 5 | 2 |

Feature Creation:

Delivery Difference = Actual Delivery Days - Expected Delivery Days

Simple Explanation:

If an order takes 5 days to arrive but was expected to arrive in 3 days, the delivery difference is 2 days.

Real-World Usage:

Delivery companies compare actual delivery times with estimated delivery times.

Business Usage:

Businesses can identify delayed deliveries and evaluate delivery performance.

AI/ML Usage:

A model can use historical delivery differences to understand patterns related to delivery performance or customer satisfaction.

Interpretation:

- Positive value: Delivery was later than expected.
- Negative value: Delivery was earlier than expected.
- Zero: Delivery matched the expected time.

Note:

```python
import pandas as pd

data = {
    "Order": ["O001", "O002", "O003", "O004"],
    "ActualDeliveryDays": [5, 2, 4, 7],
    "ExpectedDeliveryDays": [3, 4, 4, 5]
}

df = pd.DataFrame(data)

df["DeliveryDifference"] = (
    df["ActualDeliveryDays"]
    - df["ExpectedDeliveryDays"]
)

print(df)
```

## Step 3 – Implement the Concept

3. Create a DataFrame:
   - pd.DataFrame(data) converts the dictionary into a tabular structure.
   - Each row represents one order.

4. Create the new feature:
   - df["DeliveryDifference"] creates a new column.
   - df["ActualDeliveryDays"] retrieves the actual delivery duration.
   - df["ExpectedDeliveryDays"] retrieves the expected delivery duration.
   - The - operator subtracts the expected duration from the actual duration row by row.

5. Display the result:
   - print(df) displays the original features and the newly created feature.

Expected Output:

1  O002                   2                     4                  -2
2  O003                   4                     4                   0
3  O004                   7                     5                   2

Why is subtraction used?

Subtraction is appropriate because both features measure delivery duration in days. Their difference represents how early or late an order was delivered.

Before Feature Creation:
Actual and expected delivery durations are stored separately.

After Feature Creation:
DeliveryDifference represents the gap between the actual and expected delivery durations.

AI/ML Application:

The difference can help identify delivery performance patterns and may provide useful information for related prediction tasks.

Important:
If the goal is to predict delivery delay before delivery happens, actual delivery duration is not available yet. Using it as an input would cause data leakage. Use only information available at prediction time.

## Step 4 – Justify Your Feature

### Feature Name
DeliveryDifference

### Source Columns
- ActualDeliveryDays
- ExpectedDeliveryDays

### Feature Logic / Formula
DeliveryDifference = ActualDeliveryDays - ExpectedDeliveryDays

### Reason for Creation
The feature calculates the difference between actual and expected delivery durations. It helps identify whether an order was early, on time, or delayed.

### Expected Relationship with Target
If the target is a historical delivery performance category, DeliveryDifference may have a meaningful relationship with it.

For example, larger positive values indicate greater delivery delays.

### Leakage Risk
ActualDeliveryDays is only known after delivery is completed.

If the prediction is made before delivery, using actual delivery duration creates data leakage.

Use this feature only when both source values are available at the prediction time.

### Final Decision (Retain/Remove)
Retain for historical delivery analysis or prediction tasks where both values are available at prediction time.

Remove from pre-delivery prediction inputs if actual delivery duration is not yet available.

# Topic 5 – Feature / Feature

## Step 1 – Understand the Concept

Feature / Feature is a feature engineering technique in which one numerical feature is divided by another numerical feature to create a new feature.

Division helps calculate ratios and understand the relationship between two numerical values.

Why do we use Feature / Feature?

- To compare two related numerical values.
- To calculate per-unit values.
- To understand proportions and relationships.
- To create meaningful ratio-based features.

When is it useful?

- When both features have a meaningful relationship.
- When the denominator is not zero.
- When the resulting ratio has a clear business interpretation.

Simple Example:

Total Spending / Number of Purchases = Average Spending Per Purchase

Real-World Example:

Distance / Time = Average Speed

Business Example:

An e-commerce company can calculate average spending per purchase to understand customer purchasing behaviour.

AI/ML Use Case:

A machine learning model can use average spending per purchase to identify differences in customer purchasing patterns.

Important Considerations:

- Check for zero values in the denominator.
- Handle missing and invalid values.
- Understand the units of both features.
- Avoid dividing unrelated features.
- Use only information available at prediction time.

## Step 2 – Demonstrate the Concept

Consider an e-commerce company that tracks the total spending and number of purchases made by each customer.

The company wants to calculate the average spending per purchase.

Example:

| Customer | Total Spending | Number of Purchases | Average Spending |
|---|---:|---:|---:|
| C001 | 1000 | 5 | 200 |
| C002 | 2400 | 6 | 400 |
| C003 | 900 | 3 | 300 |
| C004 | 1600 | 4 | 400 |

Feature Creation:

Average Spending = Total Spending / Number of Purchases

Simple Explanation:

If a customer spends ₹1000 across 5 purchases, the average spending per purchase is ₹200.

Real-World Usage:

Online shopping platforms analyze how much customers spend on average for each purchase.

Business Usage:

Businesses can compare customer spending patterns and identify customers with higher or lower average purchase values.

AI/ML Usage:

Average spending per purchase may help a model understand customer purchasing behaviour and predict related business outcomes.

Important:

If the number of purchases is zero, division is not valid. Such cases must be handled before creating the feature.

```python
import pandas as pd
import numpy as np

data = {
    "Customer": ["C001", "C002", "C003", "C004"],
    "TotalSpending": [1000, 2400, 900, 1600],
    "NumberOfPurchases": [5, 6, 3, 4]
}

df = pd.DataFrame(data)

df["AverageSpending"] = np.where(
    df["NumberOfPurchases"] != 0,
    df["TotalSpending"] / df["NumberOfPurchases"],
    np.nan
)

print(df)
```

## Step 3 – Implement the Concept

Code Explanation:

1. Import Pandas:
   - import pandas as pd
   - Pandas is used to create and manipulate tabular data.

2. Import NumPy:
   - import numpy as np
   - NumPy provides numerical operations and functions such as np.where().

3. Create the sample data:
   - The data dictionary contains Customer, TotalSpending, and NumberOfPurchases.
   - TotalSpending represents the total amount spent by a customer.
   - NumberOfPurchases represents the number of purchases made by that customer.

4. Create a DataFrame:
   - pd.DataFrame(data) converts the dictionary into a tabular structure.
   - Each row represents one customer.

5. Create the new feature:
   - df["AverageSpending"] creates a new column.
   - df["TotalSpending"] retrieves the total spending.
   - df["NumberOfPurchases"] retrieves the purchase count.
   - The / operator divides total spending by the number of purchases.

6. Handle division by zero:
   - np.where() checks whether NumberOfPurchases is not equal to zero.
   - If the condition is true, division is performed.
   - If the condition is false, np.nan is assigned.
   - np.nan represents a missing numerical value.

7. Display the result:
   - print(df) displays the original features and the newly created feature.

Expected Output:

  Customer  TotalSpending  NumberOfPurchases  AverageSpending
0     C001           1000                  5            200.0
1     C002           2400                  6            400.0
2     C003            900                  3            300.0
3     C004           1600                  4            400.0

Why is division used?

Division is appropriate because it calculates the average spending for each purchase instead of considering only the total spending.

Before Feature Creation:
Total spending and purchase count are stored separately.

After Feature Creation:
AverageSpending represents the average amount spent per purchase.

AI/ML Application:

The feature can help a model compare customers based on their average purchase value.

Important:
In this example, np.where() selects the result for each row. For real datasets, check the denominator and handle missing values before model training.

If this feature is calculated using historical transactions, use only transactions available before the prediction time.

## Step 4 – Justify Your Feature

### Feature Name
AverageSpending

### Source Columns
- TotalSpending
- NumberOfPurchases

### Feature Logic / Formula
AverageSpending = TotalSpending / NumberOfPurchases

### Reason for Creation
The feature calculates the average amount spent per purchase. It provides more information about customer purchasing behaviour than total spending alone.

### Expected Relationship with Target
If the target is a future customer spending or purchasing outcome, AverageSpending may provide useful historical information.

Its actual relationship with the target must be checked using training and validation data.

### Potential Business Meaning

# Topic 6 – Polynomial Features

## Step 1 – Understand the Concept

Polynomial Features are created by raising existing numerical features to powers or combining them through polynomial terms.

They help machine learning models understand non-linear relationships between input features and the target.

Why do we use Polynomial Features?

- To represent non-linear patterns in data.
- To capture relationships that a simple linear model may not understand.
- To create squared and higher-degree feature values.
- To improve the ability of certain models to learn complex patterns.

When are they useful?

- When the relationship between a feature and the target is non-linear.
- When a linear model does not capture the observed pattern.
- When polynomial terms improve validation performance.

Simple Example:

If x = 3:

- x = 3
- x² = 9
- x³ = 27

Real-World Example:

A vehicle's fuel consumption may change non-linearly with speed under certain driving conditions.

Business Example:

A business may study whether advertising expenditure has a non-linear relationship with sales.

AI/ML Use Case:

Polynomial features can help a linear regression model represent non-linear patterns by adding squared or interaction terms.

Important Considerations:

- Higher-degree polynomial features can increase model complexity.
- They may cause overfitting.
- Large numerical values can produce very large polynomial values.
- Use validation data to evaluate their usefulness.
- Fit preprocessing transformations using training data only.

## Step 2 – Demonstrate the Concept

Consider a business that wants to study the relationship between advertising expenditure and sales.

The business observes that sales may not always increase at a constant rate when advertising expenditure increases.

Example:

| Advertising Spend | Advertising Spend Squared |
|---:|---:|
| 1 | 1 |
| 2 | 4 |
| 3 | 9 |
| 4 | 16 |

Feature Creation:

Advertising Spend Squared = Advertising Spend × Advertising Spend

Simple Explanation:

If advertising expenditure is 3 units, its squared feature is 9.

Real-World Usage:

Businesses can study whether increasing advertising expenditure has a non-linear relationship with sales.

Business Usage:

The squared feature may help represent patterns such as diminishing returns, where additional advertising expenditure produces smaller increases in sales.

AI/ML Usage:

A linear regression model can use both AdvertisingSpend and AdvertisingSpendSquared to represent a curved relationship.

Important:

Creating a squared feature does not automatically prove that the relationship is non-linear. Its usefulness must be evaluated using validation data.

```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures

data = {
    "AdvertisingSpend": [1, 2, 3, 4, 5]
}

df = pd.DataFrame(data)

poly = PolynomialFeatures(
    degree=2,
    include_bias=False
)

polynomial_data = poly.fit_transform(
    df[["AdvertisingSpend"]]
)

feature_names = poly.get_feature_names_out(
    ["AdvertisingSpend"]
)

polynomial_df = pd.DataFrame(
    polynomial_data,
    columns=feature_names
)

print(polynomial_df)
```

   - The data dictionary contains AdvertisingSpend.
   - It represents the advertising expenditure values used in the example.

4. Create a DataFrame:
   - pd.DataFrame(data) converts the dictionary into a table.

5. Initialize PolynomialFeatures:
   - PolynomialFeatures(degree=2, include_bias=False) creates polynomial features up to degree 2.
   - degree=2 generates squared terms and pairwise interaction terms when multiple input features are provided.
   - include_bias=False prevents the creation of an additional constant column of ones.

6. Generate polynomial features:
   - poly.fit_transform() learns the required feature structure and transforms the input data.
   - df[["AdvertisingSpend"]] selects the input column.
   - The transformed values are stored in polynomial_data.

7. Get feature names:
   - get_feature_names_out() returns the names of the generated features.
   - This helps identify the original and newly created columns.

8. Create the output DataFrame:
   - pd.DataFrame() converts the transformed numerical array into a readable table.
   - columns=feature_names assigns the generated feature names.

   - print(polynomial_df) displays the generated polynomial features.

Why is PolynomialFeatures used?

AI/ML Application:

Important:
For a real ML workflow, split the data into training and testing sets first. Fit PolynomialFeatures using training data and use transform() on testing data. Select the polynomial degree based on validation performance.

# Topic 7 – Pairwise Interactions

## Step 1 – Understand the Concept

Pairwise Interactions are new features created by combining two different input features.

They help a machine learning model understand how two features may work together to influence a target.

Why do we use Pairwise Interactions?

- To identify relationships between two features.
- To represent combined effects.
- To capture patterns that individual features may not explain.
- To provide additional information to machine learning models.

When are they useful?

- When two features may jointly influence the target.
- When the combined effect is meaningful in the business domain.
- When validation results show that the interaction improves model performance.

Simple Example:

Study Hours × Practice Hours = Study Practice Interaction

Real-World Example:

Advertising Spend × Website Traffic can represent the combined relationship between advertising activity and website visits.

Business Example:

An e-commerce company may study how product price and discount percentage work together to influence customer purchases.

AI/ML Use Case:

Pairwise interaction features can help linear models represent combined effects between input features.

Important Considerations:

- Select features that have a meaningful relationship.
- Avoid creating every possible combination without a reason.
- Too many interactions can increase model complexity.
- Check for overfitting using validation data.
- Use only information available at prediction time.

## Step 2 – Demonstrate the Concept

Consider an e-commerce business that studies the relationship between product price and discount percentage.

The business wants to understand how these two features work together.

Example:

| Product | Price | Discount Percentage | Price Discount Interaction |
|---|---:|---:|---:|
| P001 | 1000 | 10 | 10000 |
| P002 | 2000 | 20 | 40000 |
| P003 | 1500 | 10 | 15000 |
| P004 | 3000 | 30 | 90000 |

Feature Creation:

Price Discount Interaction = Price × Discount Percentage

Simple Explanation:

If a product price is ₹1000 and its discount percentage is 10, the interaction value is 10000.

Real-World Usage:

Online shopping platforms analyze product prices and discounts together to understand purchasing patterns.

Business Usage:

The interaction may help a business study whether discounts have different effects on low-priced and high-priced products.

AI/ML Usage:

A machine learning model can use this interaction feature to represent the combined effect of product price and discount percentage.

Important:

The interaction value is not the final discount amount. It is a combined numerical feature.

The actual discount amount can be calculated as:

Discount Amount = Price × Discount Percentage / 100

The interaction feature is useful only when it provides additional information beyond the original features.

```python
import pandas as pd

data = {
    "Product": ["P001", "P002", "P003", "P004"],
    "Price": [1000, 2000, 1500, 3000],
    "DiscountPercentage": [10, 20, 10, 30]
}

df = pd.DataFrame(data)

df["Price_Discount_Interaction"] = (
    df["Price"] * df["DiscountPercentage"]
)

print(df)
```

   - Each row represents one product.

4. Create the pairwise interaction feature:
   - df["Price_Discount_Interaction"] creates a new column.
   - df["Price"] retrieves the product price.
   - df["DiscountPercentage"] retrieves the discount percentage.
   - The multiplication operator combines the two features row by row.

   - print(df) displays the original features and the newly created interaction feature.

Why is multiplication used?

Multiplication creates a combined feature that represents the relationship between product price and discount percentage.
