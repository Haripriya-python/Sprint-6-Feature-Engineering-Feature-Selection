# Topic 1 – Count

## Step 1 – Understand the Concept

Count is an aggregation operation used to calculate the number of records or occurrences in a dataset.

It helps summarize how many times an event occurs or how many records belong to a particular group.

Why is Count important?

- It helps calculate the number of customer purchases.
- It helps identify frequently purchased products.
- It helps measure customer activity.
- It supports business analysis and machine learning.

When is it used?

Count is commonly used in customer analysis, product analysis, transaction analysis, and business reporting.

Simple Example:

If a customer makes 5 purchases, the purchase count is 5.

Real-World Example:

An online shopping platform counts how many orders each customer places.

Business Example:

A company identifies customers who place more orders and customers who place fewer orders.

AI/ML Use Case:

Customer purchase count can be used as an input feature in customer segmentation and churn prediction models.

Important Difference:

- Count of rows measures the number of records.
- Count of unique invoices measures the number of distinct orders.

In an online retail dataset, one invoice can contain multiple product rows. Therefore, unique invoice count is more appropriate when calculating the number of customer orders.

## Step 2 – Demonstrate the Concept

Consider an online shopping company with the following transaction records.

| Customer | Invoice | Product |
|---|---|---|
| C001 | INV101 | Laptop |
| C001 | INV101 | Mouse |
| C001 | INV102 | Keyboard |
| C002 | INV103 | Monitor |
| C002 | INV104 | Mouse |
| C003 | INV105 | Laptop |
| C003 | INV105 | Mouse |

The company wants to calculate the number of orders placed by each customer.

The calculation is:

Purchase Count = Number of Unique Invoices for Each Customer

Examples:

- C001 has 2 unique invoices.
- C002 has 2 unique invoices.
- C003 has 1 unique invoice.

Notice that invoice INV101 contains two product rows, but it represents only one order.

Business Example:

The company can identify how many orders each customer has placed and understand their purchasing activity.

AI/ML Example:

The purchase count can help a machine learning model identify differences between customers with frequent and occasional purchases.

```python
import pandas as pd

# Create a sample transaction dataset
df = pd.DataFrame({
    "Customer": [
        "C001", "C001", "C001",
        "C002", "C002",
        "C003", "C003"
    ],
    "Invoice": [
        "INV101", "INV101", "INV102",
        "INV103", "INV104",
        "INV105", "INV105"
    ],
    "Product": [
        "Laptop", "Mouse", "Keyboard",
        "Monitor", "Mouse",
        "Laptop", "Mouse"
    ]
})

# Count unique invoices for each customer
customer_count = (
    df.groupby("Customer")["Invoice"]
    .nunique()
    .reset_index(name="PurchaseCount")
)

# Display the result
print(customer_count)
```

## Step 3 – Implement the Concept

### 1. Import Pandas

`import pandas as pd`

- Imports the Pandas library.
- Pandas is used to create, group, and analyze tabular data.

### 2. Create the Sample Dataset

`df = pd.DataFrame({...})`

- Creates a DataFrame containing customer transaction details.
- Customer represents the customer ID.
- Invoice represents the order or invoice number.
- Product represents the purchased product.

### 3. Group Records by Customer

`df.groupby("Customer")`

- Groups all transaction records using the Customer column.
- Each customer's records are analyzed separately.

### 4. Select the Invoice Column

`["Invoice"]`

- Selects the Invoice column from each customer group.
- This column contains the invoice identifiers used to identify orders.

### 5. Count Unique Invoices

`.nunique()`

- Counts the distinct invoice values for each customer.
- Repeated invoice numbers are counted only once.
- This avoids counting multiple product rows from the same invoice as separate orders.

### 6. Convert the Result into a DataFrame

`.reset_index(name="PurchaseCount")`

- Converts the grouped result into a DataFrame.
- Names the calculated feature PurchaseCount.

### 7. Display the Result

`print(customer_count)`

- Displays each customer and their total number of unique orders.

### Expected Output

| Customer | PurchaseCount |
|---|---:|
| C001 | 2 |
| C002 | 2 |
| C003 | 1 |

### Why do we use this approach?

- Grouping separates records by customer.
- `nunique()` counts distinct invoices.
- It prevents multiple product rows from being treated as separate orders.
- The result converts transaction data into a useful customer-level feature.

### Important Considerations

- Use `nunique()` when counting unique orders.
- Use `count()` when counting non-missing values in a column.
- Use `size()` when counting all rows in each group, including rows with missing values.
- Define the observation period before calculating purchase counts.
- Use only transactions available before the prediction date.

# Topic 2 – Sum

## Step 1 – Understand the Concept

Sum is an aggregation operation used to calculate the total of numerical values in a dataset.

It helps combine multiple values into a single total.

Why is Sum important?

- It helps calculate total customer spending.
- It helps calculate total products sold.
- It helps measure total sales revenue.
- It supports business performance analysis.

When is it used?

Sum is commonly used in sales analysis, customer analysis, product analysis, financial reporting, and inventory management.

Simple Example:

If a customer spends ₹500, ₹700, and ₹300 on three purchases, the total spending is ₹1,500.

Real-World Example:

An online shopping company calculates the total amount spent by each customer.

Business Example:

A retail company calculates the total quantity sold for each product to understand product demand.

AI/ML Use Case:

Total customer spending can be used as an input feature in customer segmentation, customer value analysis, and churn prediction models.

## Step 2 – Demonstrate the Concept

Consider an online shopping company with the following transaction records.

| Customer | Invoice | Product | Amount |
|---|---|---|---:|
| C001 | INV101 | Laptop | 30000 |
| C001 | INV101 | Mouse | 1000 |
| C001 | INV102 | Keyboard | 2000 |
| C002 | INV103 | Monitor | 12000 |
| C002 | INV104 | Mouse | 1000 |
| C003 | INV105 | Laptop | 30000 |

The company wants to calculate the total spending of each customer.

The calculation is:

Total Customer Spending = Sum of All Purchase Amounts

Examples:

- C001: 30000 + 1000 + 2000 = ₹33,000
- C002: 12000 + 1000 = ₹13,000
- C003: 30000 = ₹30,000

Business Example:

The company can identify customers who contribute more to total sales.

AI/ML Example:

Total customer spending can help a machine learning model understand differences in customer purchasing value.

```python
import pandas as pd

# Create a sample transaction dataset
df = pd.DataFrame({
    "Customer": [
        "C001", "C001", "C001",
        "C002", "C002",
        "C003"
    ],
    "Invoice": [
        "INV101", "INV101", "INV102",
        "INV103", "INV104",
        "INV105"
    ],
    "Product": [
        "Laptop", "Mouse", "Keyboard",
        "Monitor", "Mouse", "Laptop"
    ],
    "Amount": [
        30000, 1000, 2000,
        12000, 1000, 30000
    ]
})

# Calculate total spending for each customer
customer_spending = (
    df.groupby("Customer")["Amount"]
    .sum()
    .reset_index(name="TotalSpending")
)

# Display the result
print(customer_spending)
```

# Topic 3 – Mean

## Step 1 – Understand the Concept

Mean is an aggregation operation used to calculate the average of numerical values.

It is calculated by adding all values and dividing the total by the number of values.

Why is Mean important?

- It helps calculate average customer spending.
- It helps identify average order value.
- It helps understand typical sales values.
- It supports customer and product performance analysis.

When is it used?

Mean is commonly used in customer analysis, sales analysis, product analysis, and business reporting.

Simple Example:

A customer makes three purchases worth ₹500, ₹700, and ₹300.

Mean = (500 + 700 + 300) / 3

Average Spending = ₹500

Real-World Example:

An online shopping company calculates the average amount spent by each customer per order.

Business Example:

A retail company calculates the average sales value of its products to compare product performance.

AI/ML Use Case:

Average customer spending can be used as an input feature in customer segmentation, churn prediction, and customer value analysis.

Important Difference:

- Total spending measures the sum of all purchase amounts.
- Average spending measures the average amount per purchase.

Mean can be affected by extremely high or low values, so it should be interpreted with the data distribution.

## Step 2 – Demonstrate the Concept

Consider an online shopping company with the following customer purchase records.

| Customer | Invoice | Order Amount |
|---|---|---:|
| C001 | INV101 | 1000 |
| C001 | INV102 | 2000 |
| C001 | INV103 | 3000 |
| C002 | INV104 | 1500 |
| C002 | INV105 | 2500 |
| C003 | INV106 | 4000 |

The company wants to calculate the average order value for each customer.

The calculation is:

Average Order Value = Total Order Amount / Number of Orders

Examples:

- C001: (1000 + 2000 + 3000) / 3 = ₹2,000
- C002: (1500 + 2500) / 2 = ₹2,000
- C003: 4000 / 1 = ₹4,000

Business Example:

The company can compare the average order value of different customers.

AI/ML Example:

Average order value can help a machine learning model identify customer spending patterns.

```python
import pandas as pd

# Create a sample order-level dataset
df = pd.DataFrame({
    "Customer": [
        "C001", "C001", "C001",
        "C002", "C002",
        "C003"
    ],
    "Invoice": [
        "INV101", "INV102", "INV103",
        "INV104", "INV105",
        "INV106"
    ],
    "OrderAmount": [
        1000, 2000, 3000,
        1500, 2500,
        4000
    ]
})

# Calculate average order value for each customer
customer_average = (
    df.groupby("Customer")["OrderAmount"]
    .mean()
    .reset_index(name="AverageOrderValue")
)

# Display the result
print(customer_average)
```

### Important Considerations

### Difference Between Mean and Sum

- Sum represents the total value of all orders.
- Mean represents the average value per order.
- A customer can have high total spending because of many orders, even if their average order value is low.

# Topic 4 – Median

## Step 1 – Understand the Concept

Median is an aggregation operation used to identify the middle value in a sorted set of numerical values.

If the dataset contains an even number of values, the median is calculated by taking the average of the two middle values.

Why is Median important?

- It helps identify the typical value in a dataset.
- It reduces the influence of extreme values.
- It helps understand customer spending patterns.
- It supports analysis of data with outliers.

When is it used?

Median is commonly used in customer spending analysis, salary analysis, product pricing, and business reporting.

Simple Example:

Consider the values 100, 200, 300, 400, and 500.

The middle value is 300.

Therefore, the median is 300.

Real-World Example:

A shopping company calculates the median customer spending to understand typical spending without giving too much importance to unusually large purchases.

Business Example:

A company compares the median order value across different customer groups to understand typical purchasing behavior.

AI/ML Use Case:

Median customer spending can be used as an input feature in customer segmentation and customer behavior analysis.

Important Difference:

- Mean calculates the average of all values.
- Median identifies the middle value after sorting.
- Median is generally less affected by extremely high or low values than mean.

## Step 2 – Demonstrate the Concept

Consider the following order amounts from an online shopping company.

| Customer | Order Amount |
|---|---:|
| C001 | 500 |
| C001 | 600 |
| C001 | 700 |
| C001 | 800 |

The company wants to calculate the median order value for this customer.

First, arrange the order amounts in ascending order:

500, 600, 700, 800, 10000

The middle value is 700.

Therefore:

Median Order Value = ₹700

Compare Mean and Median:

Mean = (500 + 600 + 700 + 800 + 10000) / 5 = ₹2,520

Median = ₹700

The unusually high order value increases the mean, but the median remains 700.

Business Example:

AI/ML Example:

Median-based features can help represent customer spending patterns when transaction amounts contain extreme values.

```python
import pandas as pd

# Create a sample customer order dataset
df = pd.DataFrame({
    "Customer": [
        "C001", "C001", "C001", "C001", "C001",
        "C002", "C002", "C002", "C002",
        "C003", "C003", "C003"
    ],
    "OrderAmount": [
        500, 600, 700, 800, 10000,
        1000, 1200, 1400, 1600,
        2000, 2500, 3000
    ]
})

# Calculate median order value for each customer
customer_median = (
    df.groupby("Customer")["OrderAmount"]
    .median()
    .reset_index(name="MedianOrderValue")
)

# Display the result
print(customer_median)
```

# Topic 5 – Minimum

Minimum is an aggregation operation used to identify the smallest value in a group of records.

It helps us understand the lowest transaction amount, lowest product price, minimum quantity, or minimum customer spending value.

Why do we use Minimum?
- To identify the lowest value in a dataset.
- To understand the minimum amount spent by a customer.
- To identify unusually low transaction values.
- To compare minimum values across customers or products.

When do we use Minimum?
- When analyzing customer purchasing behavior.
- When finding the lowest transaction amount.
- When identifying the cheapest product.

Simple Example:

# Topic 11 – Transaction-Level Aggregation

## Step 1 – Understand the Concept

Transaction-Level Aggregation is a technique used to summarize multiple product records belonging to the same transaction or invoice.

A single transaction may contain multiple products. Transaction-level aggregation combines those product records to calculate meaningful transaction features.

Common transaction-level features include:
- Total Transaction Amount
- Total Items Purchased
- Unique Products Count
- Average Product Price
- Maximum Product Price

Why do we use Transaction-Level Aggregation?
- To calculate the total value of each transaction.
- To understand the number of products purchased in an order.
- To summarize multiple product records into one transaction.
- To analyze transaction size and purchasing behavior.

When do we use Transaction-Level Aggregation?
- When calculating invoice totals.
- When identifying large transactions.
- When measuring the number of products in an order.
- When analyzing average transaction value.

Simple Example:
If an invoice contains three products worth ₹500, ₹700, and ₹800, the total transaction amount is ₹2,000.

Real-World Example:
An online shopping platform combines all products in a customer's cart to calculate the final invoice amount.

Business Example:
A retail company analyzes transaction totals to understand order values and purchasing patterns.

AI/ML Use Case:
Transaction-level features can help ML models predict transaction value, identify unusual transactions, and understand purchasing patterns.

Important Note:
Transaction-level aggregation must use the correct transaction identifier. In an online retail dataset, one invoice can contain multiple product rows.

## Step 2 – Demonstrate the Concept

Consider an online shopping dataset containing multiple products in different invoices.

| Invoice | StockCode | Quantity | Price |
|---------|-----------|----------|-------|
| INV101  | P101      | 2        | 500   |
| INV101  | P102      | 1        | 700   |
| INV101  | P103      | 1        | 800   |
| INV102  | P101      | 3        | 500   |
| INV102  | P104      | 2        | 300   |

Simple Example:

For Invoice INV101:

Product P101 = 2 × 500 = ₹1,000

Product P102 = 1 × 700 = ₹700

Product P103 = 1 × 800 = ₹800

Total Transaction Amount = ₹2,500

Total Items Purchased = 2 + 1 + 1 = 4

Real-World Example:
An online shopping platform calculates the total amount and number of items in each invoice.

Business Example:
A retail company compares invoice totals to understand transaction sizes and customer purchasing patterns.

AI/ML Use Case:
Transaction amount and item count can help an ML model learn transaction patterns and predict transaction values.

Expected Transaction-Level Summary:

| Invoice | TotalTransactionAmount | TotalItems | UniqueProducts |
|---------|------------------------|------------|----------------|
| INV101  | 2500                   | 4          | 3              |

```python
import pandas as pd

data = {
    "Invoice": [
        "INV101", "INV101", "INV101",
        "INV102", "INV102",
        "INV103"
    ],
    "StockCode": [
        "P101", "P102", "P103",
        "P101", "P104",
        "P102"
    ],
    "Quantity": [
        2, 1, 1,
        3, 2,
        2
    ],
    "Price": [
        500, 700, 800,
        500, 300,
        700
    ]
}

df = pd.DataFrame(data)

df["LineAmount"] = df["Quantity"] * df["Price"]

transaction_summary = (
    df.groupby("Invoice")
    .agg(
        TotalTransactionAmount=("LineAmount", "sum"),
        TotalItems=("Quantity", "sum"),
        UniqueProducts=("StockCode", "nunique"),
        AverageProductPrice=("Price", "mean"),
        MaximumProductPrice=("Price", "max")
    )
    .reset_index()
)

print(transaction_summary)
```
