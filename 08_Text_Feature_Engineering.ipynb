### Step 1: Understand the Concept – Text Length

Text Length is a basic text feature engineering technique used to measure the total number of characters in a text.

It helps convert raw text into a numerical feature that Machine Learning models can process.

For example:

Text: "Machine Learning"

Text Length: 16

Text length can help identify differences between short and long text messages, customer reviews, feedback, and descriptions.

It is useful in Natural Language Processing (NLP) tasks such as text classification and sentiment analysis.

### Step 2: Demonstrate the Concept – Text Length

Consider the following customer feedback:

1. "Good"
2. "Product is very useful"
3. "The delivery was late"

Each sentence has a different text length.

- "Good" has 4 characters.
- "Product is very useful" has 22 characters.
- "The delivery was late" has 22 characters.

By calculating text length, we can represent each text record using a numerical value.

This feature can help a Machine Learning model identify patterns related to text size.

Note: This is an illustrative sample dataset.

```python
import pandas as pd

# Create a sample text dataset
data = {
    "Customer_Feedback": [
        "Good",
        "Product is very useful",
        "The delivery was late"
    ]
}

df = pd.DataFrame(data)

# Calculate text length
df["Text_Length"] = df["Customer_Feedback"].str.len()

# Display the result
print(df)
```

### Step 1: Understand the Concept

**Sentence Count**

Sentence Count is a text feature engineering technique used to calculate the total number of sentences in a text.

A sentence usually ends with punctuation marks such as:
- Period (.)
- Question mark (?)
- Exclamation mark (!)

For example:

Text: "I like this product. It works well!"

Sentence Count: 2

It converts text into a numerical feature that can be used in Machine Learning and Natural Language Processing (NLP).

### Step 2: Demonstrate the Concept

Consider the following customer feedback:

1. "Good product."
2. "Delivery was fast. Product quality is good."
3. "I like it! Will buy again."

Sentence counts:

- "Good product." = 1 sentence
- "Delivery was fast. Product quality is good." = 2 sentences
- "I like it! Will buy again." = 2 sentences

Sentence Count helps identify how many sentences are present in each customer message.

Note: This is an illustrative sample dataset.

```python
import pandas as pd
import re

# Create a sample text dataset
data = {
    "Customer_Feedback": [
        "Good product.",
        "Delivery was fast. Product quality is good.",
        "I like it! Will buy again."
    ]
}

df = pd.DataFrame(data)

# Define a function to count sentences
def count_sentences(text):
    sentences = re.findall(r'[^.!?]+', text)
    return len([sentence for sentence in sentences if sentence.strip()])

# Calculate sentence count
df["Sentence_Count"] = df["Customer_Feedback"].apply(count_sentences)

# Display the result
print(df)
```

Note: This is an illustrative sample dataset.

```python
import pandas as pd

# Create a sample text dataset
data = {
    "Customer_Message": [
        "Hello! Welcome.",
        "Offer @ 50% OFF!!!",
        "Order #123 - Confirmed"
    ]
}

df = pd.DataFrame(data)

# Calculate the number of special characters
df["Special_Character_Count"] = df["Customer_Message"].apply(
    lambda text: sum(
        not character.isalnum() and not character.isspace()
        for character in text
    )
)

# Display the result
print(df)
```

### Step 1: Understand the Concept

**Keyword Count**

Keyword Count is a text feature engineering technique used to count how many times specific keywords appear in a text.

Keywords are important words related to a particular task or business requirement.

For example:

Text: "The product is good and the service is good"

Keywords: ["good", "service"]

Keyword Count: 3

It helps identify the presence and frequency of important words in raw text.

This feature is useful in sentiment analysis, spam detection, and text classification.

### Step 2: Demonstrate the Concept

Consider the following customer feedback:

1. "Good product and good service"
2. "Bad delivery and poor quality"
3. "Excellent product quality"

Selected Keywords:

["good", "bad", "poor", "excellent"]

Keyword Counts:

- First feedback = 2
- Second feedback = 2
- Third feedback = 1

Keyword Count helps measure how frequently selected keywords appear in customer feedback.

Note: This is an illustrative sample dataset.

```python
import pandas as pd

# Create a sample text dataset
data = {
    "Customer_Feedback": [
        "Good product and good service",
        "Bad delivery and poor quality",
        "Excellent product quality"
    ]
}

df = pd.DataFrame(data)

# Define important keywords
keywords = ["good", "bad", "poor", "excellent"]

# Count selected keywords in each feedback
def count_keywords(text):
    words = text.lower().split()
    return sum(words.count(keyword) for keyword in keywords)

# Create keyword count feature
df["Keyword_Count"] = df["Customer_Feedback"].apply(count_keywords)

# Display the result
print(df)
```

### Step 1: Understand the Concept

Bag of Words is a text vectorization technique used to convert raw text into numerical features based on word frequency.

It creates a vocabulary containing unique words from the given text documents.

Each document is represented by a numerical vector that shows how many times each vocabulary word appears.

For example:

Text 1: "I like Python"

Text 2: "I like Machine Learning"

Vocabulary:
["I", "like", "Python", "Machine", "Learning"]

Text 1 Vector: [1, 1, 1, 0, 0]

Text 2 Vector: [1, 1, 0, 1, 1]

Bag of Words helps Machine Learning models process text data as numerical values.

It is commonly used in Natural Language Processing (NLP), text classification, and spam detection.

### Step 1: Understand the Concept

**TF-IDF (Term Frequency – Inverse Document Frequency)**

TF-IDF is a text vectorization technique used to convert raw text into numerical features based on the importance of words.

It gives higher importance to words that are frequent in a particular document but less common across all documents.

TF-IDF has two main components:

1. Term Frequency (TF)
   - Measures how frequently a word appears in a document.

2. Inverse Document Frequency (IDF)
   - Measures how rare or common a word is across all documents.

For example:

If the word "excellent" appears frequently in one customer review but rarely in other reviews, it may receive a higher TF-IDF score.

TF-IDF helps Machine Learning models identify important words in text data.

It is commonly used in NLP, search engines, text classification, and sentiment analysis.

### Step 2: Demonstrate the Concept

Consider the following customer feedback:

1. "Good product quality"
2. "Good service quality"
3. "Excellent product"

In these documents:

- The word "good" appears in multiple documents.
- The word "quality" appears in multiple documents.
- The word "excellent" appears in only one document.

TF-IDF gives importance to words based on their frequency in a document and their distribution across all documents.

Common words across many documents may receive lower importance.

Less common words that are meaningful to a document may receive higher importance.

Note: This is an illustrative sample dataset.

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# Create sample text documents
documents = [
    "Good product quality",
    "Good service quality",
    "Excellent product"
]

# Create a TF-IDF vectorizer
vectorizer = TfidfVectorizer()

# Convert text into TF-IDF numerical features
tfidf_matrix = vectorizer.fit_transform(documents)

# Get vocabulary words
feature_names = vectorizer.get_feature_names_out()

# Convert the result into a DataFrame
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=feature_names
)

# Display the TF-IDF features
print("Vocabulary:")
print(feature_names)

print("\nTF-IDF Matrix:")
print(tfidf_df.round(3))
```

### Step 1: Understand the Concept

**N-Grams**

N-Grams is a text feature engineering technique used to identify consecutive sequences of words in a text.

An N-Gram represents a group of N consecutive words.

Types of N-Grams:

1. Unigram (N=1)
   - Considers one word at a time.
   - Example: "Machine Learning"
   - Features: Machine, Learning

2. Bigram (N=2)
   - Considers two consecutive words.
   - Example: "Machine Learning"
   - Feature: Machine Learning

3. Trigram (N=3)
   - Considers three consecutive words.
   - Example: "I love Machine Learning"
   - Feature: I love Machine, love Machine Learning

N-Grams help capture word sequences and relationships that individual words may not represent.

They are useful in Natural Language Processing (NLP), text classification, and sentiment analysis.

### Step 2: Demonstrate the Concept

Consider the following customer feedback:

Text: "I love this product"

Unigram:
["I", "love", "this", "product"]

Bigram:
["I love", "love this", "this product"]

Trigram:
["I love this", "love this product"]

N-Grams help identify combinations of consecutive words.

For example:

- "not good" can express negative sentiment.
- "very good" can express positive sentiment.

These word combinations provide more context than considering individual words alone.

Note: This is an illustrative sample dataset.

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# Create sample text documents
documents = [
    "I love this product",
    "This product is very good",
    "The service is not good"
]

# Create a Bigram vectorizer
vectorizer = CountVectorizer(ngram_range=(1, 2))

# Convert text into N-Gram features
ngram_matrix = vectorizer.fit_transform(documents)

# Get N-Gram feature names
feature_names = vectorizer.get_feature_names_out()

# Convert the result into a DataFrame
ngram_df = pd.DataFrame(
    ngram_matrix.toarray(),
    columns=feature_names
)

# Display the N-Gram features
print("N-Gram Vocabulary:")
print(feature_names)

print("\nN-Gram Feature Matrix:")
print(ngram_df)
```
