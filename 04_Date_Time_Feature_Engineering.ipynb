# Topic 1 – Year

## Step 1 – Understand the Concept

Year Extraction is a date feature engineering technique used to extract the year from a date or timestamp column.

For example:

InvoiceDate = 2024-08-15

Extracted Year = 2024

The year provides information about when an event or transaction occurred.

**Why is Year Extraction used?**

- It identifies the year of a transaction or event.
- It helps analyze yearly business trends.
- It helps compare business performance across different years.
- It provides useful time-related information for Machine Learning models.

**When should it be used?**

Year extraction can be useful for:

- Purchase transactions
- Employee joining dates
- Customer registration dates
- Order delivery dates
- Financial records

**Real-World Example:**

An e-commerce company can extract the purchase year from transaction dates to compare yearly sales performance.

**AI/ML Use Case:**

A Machine Learning model can use purchase year to identify changes in customer purchasing behavior over time.

**Important Note:**

Year values should be interpreted based on the prediction problem. Historical years can be useful, but future information must never be used to predict past outcomes.

## Step 2 – Demonstrate the Concept

**Simple Example:**

Consider these transaction dates:

- 2023-05-10
- 2024-08-15
- 2025-01-20

The extracted years are:

- 2023
- 2024
- 2025

**Real-World Example:**

An online shopping platform records the date of every customer purchase. Extracting the year helps identify when each purchase occurred.

**Business Example:**

A retail company can compare sales across 2023, 2024, and 2025 to understand yearly business growth.

**AI/ML Use Case:**

A Machine Learning model can use the purchase year as a feature to identify historical trends and changes in purchasing patterns.

**Why is it useful?**

The original timestamp contains several date and time components. Extracting the year provides a separate feature that can be analyzed independently.

```python
import pandas as pd

# Create a sample transaction dataset
df = pd.DataFrame({
    "Invoice": ["INV001", "INV002", "INV003", "INV004"],
    "InvoiceDate": [
        "2023-05-10",
        "2024-08-15",
        "2025-01-20",
        "2025-11-07"
    ]
})

# Convert InvoiceDate into datetime format
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

# Extract year from InvoiceDate
df["PurchaseYear"] = df["InvoiceDate"].dt.year

# Display the result
print(df)
```

## Step 3 – Implement the Concept

**1. Importing the Library**

`import pandas as pd`

Pandas is used to create and manipulate the transaction dataset.

**2. Creating the Dataset**

The DataFrame contains:

- `Invoice`: Unique transaction identification number.
- `InvoiceDate`: Date on which the transaction occurred.

**3. Converting the Date Column**

`pd.to_datetime(df["InvoiceDate"])`

This converts the InvoiceDate column into Pandas datetime format.

Datetime conversion is necessary because date extraction operations require proper date and time data types.

**4. Extracting the Year**

`df["InvoiceDate"].dt.year`

- `.dt` provides access to datetime properties.
- `.year` extracts the year component from each date.

**5. Creating a New Feature**

`df["PurchaseYear"]`

This stores the extracted year in a new column.

**Expected Output:**

| InvoiceDate | PurchaseYear |
|---|---:|
| 2023-05-10 | 2023 |
| 2024-08-15 | 2024 |
| 2025-01-20 | 2025 |
| 2025-11-07 | 2025 |

**Why was this feature created?**

The original InvoiceDate column contains complete date information. The new feature extracts only the year for easier analysis.

**ML Consideration:**

Use the purchase year only when it is available at the prediction time. For future prediction tasks, ensure that the feature does not contain information from dates after the prediction point.

## Step 4 – Justify Your Feature

**Feature Name:** PurchaseYear

**Source Columns:** InvoiceDate

**Feature Logic / Formula:**

PurchaseYear = Year component extracted from InvoiceDate.

**Reason for Creation:**

To identify the year in which a customer transaction occurred.

**Expected Relationship with Target:**

Purchase year may show changes in customer purchasing behavior or sales patterns over time. Its relationship with the target must be verified using actual data.

**Potential Business Meaning:**

The feature helps businesses compare yearly sales performance and understand historical purchasing trends.

**Potential ML Value:**

It provides historical time-related information that may help a Machine Learning model identify yearly patterns.

**Leakage Risk:**

Using a transaction date that occurs after the prediction date can introduce future information leakage.

Only use dates that are available at the time of prediction.

**Final Decision:** Retain

**Justification:**

Retain PurchaseYear when historical year information is relevant to the business problem and available before prediction.

# Topic 2 – Month

## Step 1 – Understand the Concept

Month is a date and time feature that represents the month in which an event or transaction occurs.

A year contains 12 months, represented by numbers from 1 to 12.

In Python Pandas, we can extract the month from a date column using `.dt.month`.

Why is Month useful?
- To understand monthly sales patterns.
- To identify high-demand and low-demand months.
- To analyze customer purchasing behaviour.
- To help predict monthly business activity.

When do we use Month?
- When analyzing sales across different months.
- When identifying seasonal purchasing patterns.
- When building machine learning models that use transaction dates.

Real-world example:
A shopping website may receive more orders during November and December because of holiday shopping.

Business example:
An e-commerce company can compare monthly revenue to identify which months generate more sales.

AI/ML use case:
A machine learning model can use the purchase month to identify monthly patterns and support sales forecasting.

Important:
Month numbers are not continuous measurements. December (12) and January (1) are adjacent months in the calendar, even though their numeric difference is 11.

## Step 2 – Demonstrate the Concept

Consider an e-commerce company that records customer purchases on different dates.

Example:

| Invoice | InvoiceDate | Purchase Month |
|---|---|---|
| INV001 | 2023-05-10 | 5 |
| INV002 | 2024-08-15 | 8 |
| INV003 | 2025-01-20 | 1 |
| INV004 | 2025-11-07 | 11 |

Simple example:
If a customer purchases a product on 15 August, the purchase month is 8.

Real-world example:
A customer may purchase more products during December because of holiday offers.

Business example:
A company can compare monthly transactions to understand which months have higher customer demand.

AI/ML example:
A sales prediction model can use the purchase month to learn whether certain months are associated with higher or lower sales.

```python
import pandas as pd

df = pd.DataFrame({
    "Invoice": ["INV001", "INV002", "INV003", "INV004"],
    "InvoiceDate": [
        "2023-05-10",
        "2024-08-15",
        "2025-01-20",
        "2025-11-07"
    ]
})

df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

df["PurchaseMonth"] = df["InvoiceDate"].dt.month

df["PurchaseMonthName"] = df["InvoiceDate"].dt.month_name()

print(df)
```

## Step 3 – Implement the Concept

### Code Explanation

**1. Import Pandas**

`import pandas as pd`

- Imports the Pandas library.
- Pandas is used to create and manipulate tabular data.
- It provides date and time operations for extracting information from date columns.

**2. Create the DataFrame**

`df = pd.DataFrame({...})`

- Creates a sample DataFrame containing invoice numbers and invoice dates.
- The Invoice column identifies each transaction.
- The InvoiceDate column stores the transaction date.

**3. Convert the Date Column**

`df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])`

- Converts the InvoiceDate column into Pandas datetime format.
- Date strings must be converted before using the `.dt` accessor.
- This allows us to extract individual date components.

**4. Extract the Month Number**

`df["PurchaseMonth"] = df["InvoiceDate"].dt.month`

- Extracts the month number from each invoice date.
- The result ranges from 1 to 12.
- For example, August becomes 8 and December becomes 12.
- Creates a new feature called PurchaseMonth.

**5. Extract the Month Name**

`df["PurchaseMonthName"] = df["InvoiceDate"].dt.month_name()`

- Extracts the full month name.
- For example, 8 becomes August.
- This column makes the output easier for humans to understand.
- It is included for readability and is not required as a separate model feature when using the numeric month.

**6. Display the Result**

`print(df)`

- Displays the original columns and newly created month features.
- Helps us verify that the month extraction worked correctly.

### Expected Output

| Invoice | InvoiceDate | PurchaseMonth | PurchaseMonthName |
|---|---|---|---|
| INV001 | 2023-05-10 | 5 | May |
| INV002 | 2024-08-15 | 8 | August |
| INV003 | 2025-01-20 | 1 | January |
| INV004 | 2025-11-07 | 11 | November |

### Why Do We Use This Method?

- `.dt.month` extracts the month number directly from a datetime column.
- It avoids manually checking each date.
- It helps create useful date-based features for data analysis and machine learning.

### AI/ML Application

PurchaseMonth can help a model learn monthly purchasing patterns.

However, month is a cyclical feature. For models that need to understand the calendar cycle, it may be represented using sine and cosine transformations so December and January are treated as nearby months.

## Step 4 – Justify Your Feature

### Feature Justification

**1. Feature Name:**
PurchaseMonth

**2. Source Columns:**
InvoiceDate

**3. Feature Logic / Formula:**
Extract the month number from InvoiceDate using:

PurchaseMonth = InvoiceDate.dt.month

**4. Reason for Creation:**
To identify the month in which each transaction occurred and analyze monthly purchasing patterns.

**5. Expected Relationship with Target:**
If the target is sales amount, PurchaseMonth may show different sales patterns across months. The relationship must be checked using historical data rather than assumed.

**6. Potential Business Meaning:**
Helps businesses identify high-sales months, low-demand periods, and possible seasonal purchasing patterns.

**7. Potential ML Value:**
Can provide useful calendar information for sales prediction and customer purchase analysis.

**8. Leakage Risk:**
The month is available from the transaction date, so extracting it does not automatically create leakage. However, the transaction date must be available at the prediction time. Future transaction information must not be used to predict past events.

**9. Final Decision:**
Retain – PurchaseMonth can provide useful information about monthly business and customer purchasing patterns. Its value should be validated using historical data and the intended prediction task.

When do we use Quarter?

The calculation is:

Purchase Frequency = Number of Purchases During the Observation Period

```python
import pandas as pd

# Create a sample customer transaction dataset
df = pd.DataFrame({
    "Customer": [
        "C001", "C001", "C001",
        "C002", "C002",
        "C003",
        "C004", "C004", "C004", "C004"
    ],
    "InvoiceDate": [
        "2025-10-02", "2025-10-10", "2025-10-20",
        "2025-10-05", "2025-10-15",
        "2025-10-08",
        "2025-10-01", "2025-10-07",
        "2025-10-14", "2025-10-21"
    ]
})

# Convert invoice dates into datetime format
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

# Define the observation period
start_date = pd.Timestamp("2025-10-01")
reference_date = pd.Timestamp("2025-10-31")

# Select transactions within the observation period
period_df = df[
    (df["InvoiceDate"] >= start_date) &
    (df["InvoiceDate"] <= reference_date)
].copy()

# Calculate purchase frequency for each customer
frequency_df = (
    period_df.groupby("Customer")
    .size()
    .reset_index(name="PurchaseFrequency")
)

# Display the result
print(frequency_df)
```

# Topic 16 – Seasonality

## Step 1 – Understand the Concept

Seasonality refers to repeated patterns in data that occur at regular time intervals.

In business data, seasonal patterns may appear during specific months, days, weeks, or seasons of the year.

Why is Seasonality important?

- It helps identify repeated purchasing patterns.
- It helps businesses understand seasonal demand.
- It supports sales and demand forecasting.
- It helps businesses plan inventory and marketing activities.

When is it used?

Seasonality is commonly used in retail sales analysis, demand forecasting, inventory planning, and customer purchase behavior analysis.

Simple Example:

An ice cream shop may experience higher sales during summer and lower sales during winter.

Real-World Example:

Online shopping platforms may experience increased purchases during festival seasons and special sale events.

Business Example:

A clothing store may observe higher sales of winter clothing during colder months and summer clothing during warmer months.

AI/ML Use Case:

Seasonal features can be used in machine learning models for sales forecasting and demand prediction.

Month, day of week, and seasonal indicators can help models identify repeated patterns in historical data.

```python
import pandas as pd

# Create a sample transaction dataset
df = pd.DataFrame({
    "InvoiceDate": [
        "2024-01-15",
        "2024-03-20",
        "2024-06-10",
        "2024-09-15",
        "2024-11-20",
        "2024-12-10",
        "2025-01-12",
        "2025-03-18",
        "2025-06-15",
        "2025-09-10",
        "2025-11-15",
        "2025-12-20"
    ],
    "Sales": [
        12000, 15000, 25000, 17000,
        28000, 35000, 13000, 16000,
        26000, 18000, 30000, 38000
    ]
})

# Convert invoice dates into datetime format
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

# Extract month from invoice date
df["PurchaseMonth"] = df["InvoiceDate"].dt.month

# Extract year from invoice date
df["PurchaseYear"] = df["InvoiceDate"].dt.year

# Create a seasonal indicator for the November-December period
df["PeakSeasonIndicator"] = (
    df["PurchaseMonth"].isin([11, 12])
).astype(int)

# Calculate average sales for each month
monthly_sales = (
    df.groupby("PurchaseMonth")["Sales"]
    .mean()
    .reset_index(name="AverageSales")
)

# Display the transaction dataset
print("Transaction Data:")
print(df)

# Display monthly average sales
print("\nMonthly Average Sales:")
print(monthly_sales)
```
