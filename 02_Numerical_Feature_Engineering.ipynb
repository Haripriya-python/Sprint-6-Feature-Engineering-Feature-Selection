# Topic 1 – Mathematical Transformations

## Step 1 – Understand the Concept

### What are Mathematical Transformations?

Mathematical Transformations are operations performed on existing numerical features to create new features or change their values into more meaningful representations.

These transformations help represent relationships between numerical columns and extract useful information for Machine Learning.

### Why are Mathematical Transformations Used?

- To create meaningful features from existing columns.
- To represent business relationships mathematically.
- To identify patterns in numerical data.
- To make data more suitable for Machine Learning algorithms.
- To improve the usefulness of input features.

### Common Mathematical Operations

1. Addition: Combines two or more numerical values.
2. Subtraction: Calculates the difference between values.
3. Multiplication: Calculates the product of values.
4. Division: Calculates the relationship between values.

Other mathematical transformations include ratios, percentages, differences, and aggregations.

### Example

In an e-commerce dataset:

Quantity = 5

Price = 200

Total Amount = Quantity × Price

Total Amount = 1000

The new feature represents the total value of the purchased items.

### When Should Mathematical Transformations Be Used?

They should be used when existing numerical columns can be combined or compared to represent meaningful business information.

The transformation must have a clear purpose and should be appropriate for the prediction task.

## Step 2 – Demonstrate the Concept

### 1. Simple Example

Consider a shopping bill:

- Product A Price = 100
- Product B Price = 200

Total Bill = 100 + 200 = 300

Addition combines the two product prices to calculate the total bill.

### 2. Real-World Example

An e-commerce company stores the quantity and price of each purchased product.

It can multiply these values to calculate the transaction line amount.

Formula:

Line Amount = Quantity × Price

### 3. Business Example

A retail company wants to calculate the total value of customer purchases.

Mathematical transformations can help calculate:

- Total Purchase Amount
- Average Purchase Value
- Discount Amount
- Remaining Stock
- Purchase Quantity Difference

These features help the business understand sales and customer behaviour.

### 4. AI/ML Use Case

In a sales prediction model, numerical transformations can create useful input features such as:

- Total Purchase Amount
- Average Purchase Value
- Discount Percentage

These features may help the model identify purchasing patterns.

### When and Why Should We Use Them?

Use mathematical transformations when the relationship between existing numerical columns has a meaningful business or analytical interpretation.

Avoid creating features without understanding their meaning or checking whether they are available at prediction time.

```python
import pandas as pd

# Create a sample e-commerce dataset
data = {
    "Invoice": ["INV001", "INV002", "INV003", "INV004", "INV005"],
    "Quantity": [2, 5, 3, 4, 6],
    "Price": [100, 200, 150, 250, 120],
    "Discount": [10, 20, 15, 25, 12],
    "ShippingCharge": [30, 40, 25, 35, 45]
}

df = pd.DataFrame(data)

# Apply mathematical transformations
df["GrossAmount"] = df["Quantity"] * df["Price"]

df["NetAmount"] = df["GrossAmount"] - df["Discount"]

df["FinalAmount"] = df["NetAmount"] + df["ShippingCharge"]

# Display the original and transformed features
print(df)
```

## Step 3 – Implement the Concept

### Detailed Code Explanation

**1. Importing Pandas**

`import pandas as pd`

Pandas is used to create and manipulate tabular data.

**2. Creating the Sample Dataset**

The dataset contains five columns:

- Invoice: Transaction identifier.
- Quantity: Number of products purchased.
- Price: Price per product.
- Discount: Discount amount applied to the transaction line.
- ShippingCharge: Shipping charge for the transaction line.

`pd.DataFrame(data)` converts the dictionary into a DataFrame.

**3. Creating GrossAmount**

`df["Quantity"] * df["Price"]` multiplies the quantity by the unit price.

Formula:

Gross Amount = Quantity × Price

Example:

Quantity = 2

Price = 100

Gross Amount = 2 × 100 = 200

The result is stored in a new column named GrossAmount.

**When to Use:** Use multiplication when two numerical values combine to represent a meaningful quantity, such as quantity multiplied by unit price.

**Why to Use:** It helps calculate the total value before discounts and other adjustments.

**4. Creating NetAmount**

`df["GrossAmount"] - df["Discount"]` subtracts the discount from the gross amount.

Formula:

Net Amount = Gross Amount - Discount

Example:

Gross Amount = 200

Discount = 10

Net Amount = 200 - 10 = 190

The result is stored in NetAmount.

**When to Use:** Use subtraction when calculating a remaining value or the difference between two numerical amounts.

**Why to Use:** It represents the transaction value after applying a discount.

**5. Creating FinalAmount**

`df["NetAmount"] + df["ShippingCharge"]` adds the shipping charge to the net amount.

Formula:

Final Amount = Net Amount + Shipping Charge

Example:

Net Amount = 190

Shipping Charge = 30

Final Amount = 190 + 30 = 220

The result is stored in FinalAmount.

**When to Use:** Use addition when multiple charges or values must be combined.

**Why to Use:** It represents the final transaction amount under the assumptions used in this example.

**6. Displaying the Result**

`print(df)` displays the original columns and the newly created mathematical features.

### Expected Output

| Invoice | GrossAmount | NetAmount | FinalAmount |
|---|---:|---:|---:|
| INV001 | 200 | 190 | 220 |
| INV002 | 1000 | 980 | 1020 |
| INV003 | 450 | 435 | 460 |
| INV004 | 1000 | 975 | 1010 |
| INV005 | 720 | 708 | 753 |

### How Does This Help Machine Learning?

- GrossAmount represents the value before discount.
- NetAmount represents the value after discount.
- FinalAmount represents the amount after adding shipping.

These features may provide useful information for sales analysis and prediction.

**Important:** The example assumes Discount and ShippingCharge are transaction-line amounts. In a real dataset, verify whether these values are stored per product, per invoice, or per order before applying the formulas.

- Reason for Creation: Calculates the transaction value before discount.

### Feature 3: FinalAmount

### Final Mathematical Transformation Decision

Mathematical transformations help convert existing numerical columns into meaningful features.

- Whether the feature has business value.

# Topic 2 – Addition

## Step 1 – Understand the Concept

### What is Addition in Feature Engineering?

Addition is a mathematical operation used to combine two or more numerical features to create a new feature.

It helps represent the combined value of different numerical columns.

### Formula

New Feature = Feature A + Feature B

### Why is Addition Used?

- To combine multiple numerical values.
- To calculate total amounts.
- To represent combined business measurements.
- To create meaningful input features for Machine Learning.

### Examples

1. Product Price + Shipping Charge = Total Cost
2. Basic Salary + Allowance = Gross Salary
3. Mathematics Marks + Science Marks = Combined Marks
4. Product Sales + Service Revenue = Total Revenue

### When Should Addition Be Used?

Addition should be used when two or more numerical columns represent values that can meaningfully be combined.

The values should have compatible units and business meanings.

For example, adding product price and shipping charge is meaningful when both represent monetary amounts in the same currency.

### Important Consideration

Adding unrelated numerical columns may create a feature that has no meaningful interpretation.

Always understand the source columns before applying addition.

## Step 2 – Demonstrate the Concept

### 1. Simple Example

Imagine purchasing two products:

- Product A = 100
- Product B = 200

Total Product Cost = 100 + 200 = 300

Addition combines the two product values into one total.

### 2. Real-World Example

An e-commerce company calculates the total customer payment.

- Product Amount = 1,000
- Shipping Charge = 50

Total Amount = 1,000 + 50 = 1,050

### 3. Business Example

A retail company wants to calculate total revenue from different product categories.

It can combine:

- Electronics Revenue
- Clothing Revenue
- Grocery Revenue

This creates a Total Revenue feature.

### 4. AI/ML Use Case

In an e-commerce prediction model, a combined monetary feature may help represent the total transaction value.

The model may use this feature to understand customer spending behaviour.

### When and Why Should We Use Addition?

Use addition when separate numerical values contribute to one meaningful total.

It is useful when the combined value represents a business measurement that may help explain the target variable.

```python
import pandas as pd

# Create a sample e-commerce dataset
data = {
    "Invoice": ["INV001", "INV002", "INV003", "INV004", "INV005"],
    "ProductAmount": [500, 800, 600, 1000, 750],
    "ShippingCharge": [40, 50, 30, 60, 45],
    "PackagingCharge": [10, 15, 10, 20, 15]
}

df = pd.DataFrame(data)

# Create a feature by adding product and shipping amounts
df["AmountWithShipping"] = (
    df["ProductAmount"] + df["ShippingCharge"]
)

# Create a feature by adding all transaction charges
df["TotalAmount"] = (
    df["ProductAmount"]
    + df["ShippingCharge"]
    + df["PackagingCharge"]
)

# Display the original and newly created features
print(df)
```

## Step 3 – Implement the Concept

### Detailed Code Explanation

- PackagingCharge: Packaging cost.

`pd.DataFrame(data)` converts the dictionary into a DataFrame.

**3. Creating AmountWithShipping**

`df["ProductAmount"] + df["ShippingCharge"]` adds the product amount and shipping charge for each row.

Formula:

Amount With Shipping = Product Amount + Shipping Charge

Example:

Product Amount = 500

Shipping Charge = 40

Amount With Shipping = 500 + 40 = 540

The calculated values are stored in a new column named AmountWithShipping.

**When to Use:** Use this calculation when the product cost and shipping charge together represent the amount before other charges.

**Why to Use:** It provides a combined monetary value that may be useful for transaction analysis.

**4. Creating TotalAmount**

The expression:

`df["ProductAmount"] + df["ShippingCharge"] + df["PackagingCharge"]`

adds all three transaction amounts.

Formula:

Total Amount = Product Amount + Shipping Charge + Packaging Charge

Example:

Product Amount = 500

Shipping Charge = 40

Packaging Charge = 10

Total Amount = 500 + 40 + 10 = 550

The result is stored in TotalAmount.

**When to Use:** Use this calculation when all the charges contribute to the same final transaction amount.

**Why to Use:** It combines multiple cost components into one meaningful feature.

**5. Displaying the Dataset**

`print(df)` displays the original columns and the newly created features.

### Expected Output

| Invoice | AmountWithShipping | TotalAmount |
|---|---:|---:|
| INV001 | 540 | 550 |
| INV002 | 850 | 865 |
| INV003 | 630 | 640 |
| INV004 | 1060 | 1080 |
| INV005 | 795 | 810 |

### How Does This Help Machine Learning?

- AmountWithShipping represents product cost plus shipping.
- TotalAmount represents the combined transaction charges.
- These features may help explain customer spending and transaction values.

**Important:** These calculations assume all amounts belong to the same transaction and currency. Verify the actual business meaning of each column before combining them.

## Step 4 – Justify Your Feature

### Feature 1: AmountWithShipping

- Feature Name: AmountWithShipping
- Source Columns: ProductAmount and ShippingCharge
- Feature Logic / Formula: ProductAmount + ShippingCharge
- Reason for Creation: Combines product cost and shipping charge.
- Expected Relationship with Target: May be related to transaction value or customer spending.
- Potential Business Meaning: Represents product cost including shipping.
- Potential ML Value: May provide useful information for transaction analysis.
- Leakage Risk: High if the target is calculated directly from this feature or if shipping information is unavailable at prediction time.
- Final Decision: Retain when appropriate for the prediction task.

### Feature 2: TotalAmount

- Feature Name: TotalAmount
- Source Columns: ProductAmount, ShippingCharge, and PackagingCharge
- Feature Logic / Formula: ProductAmount + ShippingCharge + PackagingCharge
- Reason for Creation: Combines all transaction charges into one value.
- Expected Relationship with Target: May be related to total transaction value.
- Potential Business Meaning: Represents the combined transaction amount.
- Potential ML Value: May help understand overall customer spending.
- Leakage Risk: High if the target is the same total transaction amount or is directly calculated from this feature.
- Final Decision: Retain only when appropriate for the prediction task.

### Final Addition Feature Decision

Addition is useful when multiple numerical values contribute to a meaningful total.

Before retaining an addition-based feature:

- Check that the source columns have compatible units.
- Verify that the values belong to the same transaction or observation.
- Confirm that the calculation matches the business definition.
- Check whether the information is available at prediction time.
- Evaluate whether the new feature provides useful information.

Do not use addition simply because two numerical columns are available.

# Topic 3 – Subtraction

## Step 1 – Understand the Concept

Subtraction is a mathematical operation used to calculate the difference between two numerical values.

In Feature Engineering, subtraction helps create meaningful features by finding the remaining amount, reduction, balance, or gap between two values.

Formula:

New Feature = Value A - Value B

When to Use:
- When calculating the remaining value after a deduction.
- When finding the difference between an original value and a reduced value.
- When calculating available balance or remaining stock.

Why to Use:
- To represent business information more clearly.
- To identify reductions and remaining quantities.
- To create useful numerical features for Machine Learning models.

Simple Example:
If the original price is ₹1,000 and the discount is ₹200:

Price After Discount = 1,000 - 200 = ₹800

Real-World Example:
A customer has ₹5,000 in their account and spends ₹1,500. The remaining balance is ₹3,500.

Business Example:
An e-commerce company has 100 products in stock and sells 25 products. The remaining stock is 75.

AI/ML Use Case:
An e-commerce model can use the remaining stock feature to understand product availability and support inventory-related predictions.

Subtraction is useful when both values have compatible units and represent information available at the time of prediction.

## Step 2 – Demonstrate the Concept

Consider an e-commerce business that sells electronic products.

The business maintains the original product price, discount amount, available stock, and sold quantity.

Example Dataset:

| Invoice | OriginalPrice | DiscountAmount | AvailableStock | SoldQuantity |
|---------|---------------|----------------|----------------|--------------|
| INV001  | 2000          | 200             | 100            | 25           |
| INV002  | 1500          | 150             | 80             | 20           |
| INV003  | 3000          | 500             | 60             | 15           |

Feature 1: PriceAfterDiscount

Formula:
PriceAfterDiscount = OriginalPrice - DiscountAmount

For INV001:
2000 - 200 = 1800

Feature 2: RemainingStock

Formula:
RemainingStock = AvailableStock - SoldQuantity

For INV001:
100 - 25 = 75

Why Are These Features Useful?

- PriceAfterDiscount represents the actual selling price after the discount.
- RemainingStock represents the quantity left after sales.
- These features help businesses understand pricing and inventory conditions.

AI/ML Application:

- PriceAfterDiscount may help predict customer purchase behaviour.
- RemainingStock may help identify products that need restocking.

Important:
The source values must be measured at the correct prediction time. Using future sales or inventory information can introduce data leakage.

```python
import pandas as pd

# Create a sample e-commerce dataset
data = {
    "Invoice": ["INV001", "INV002", "INV003"],
    "OriginalPrice": [2000, 1500, 3000],
    "DiscountAmount": [200, 150, 500],
    "AvailableStock": [100, 80, 60],
    "SoldQuantity": [25, 20, 15]
}

df = pd.DataFrame(data)

# Create a feature using subtraction
df["PriceAfterDiscount"] = (
    df["OriginalPrice"] - df["DiscountAmount"]
)

# Create another feature using subtraction
df["RemainingStock"] = (
    df["AvailableStock"] - df["SoldQuantity"]
)

# Display the updated dataset
print(df)
```

import pandas as pd

df["PriceAfterDiscount"] = (

)

What does this code do?

Why is this feature created?

It represents the product price after applying the discount.

Example:

2000 - 200 = 1800

The resulting value is stored in a new column named PriceAfterDiscount.

4. Creating RemainingStock

df["RemainingStock"] = (
    df["AvailableStock"] - df["SoldQuantity"]
)

What does this code do?

It subtracts the sold quantity from the available stock.

Why is this feature created?

It represents the remaining quantity after the recorded sales.

Example:

100 - 25 = 75

The resulting value is stored in a new column named RemainingStock.

print(df)

How Does This Help Machine Learning?

- PriceAfterDiscount can provide pricing information for customer purchase predictions.

These features are useful only when their source values are available at the intended prediction time.

## Step 4 – Justify Your Feature

### Feature 1: PriceAfterDiscount

- Feature Name: PriceAfterDiscount
- Source Columns: OriginalPrice, DiscountAmount
- Feature Logic / Formula: OriginalPrice - DiscountAmount
- Reason for Creation: To calculate the actual product price after applying a discount.
- Expected Relationship with Target: It may influence customer purchase behaviour. The relationship depends on the target variable and customer preferences.
- Potential Business Meaning: Shows the effective selling price of a product.
- Potential ML Value: May help a model understand the relationship between product pricing and purchasing behaviour.
- Leakage Risk: Low when the price and discount are known before the prediction. High if the values are recorded after the prediction event or directly reveal the target.
- Final Decision: Retain, after verifying the timing and meaning of the source values.

### Feature 2: RemainingStock

- Feature Name: RemainingStock
- Source Columns: AvailableStock, SoldQuantity
- Feature Logic / Formula: AvailableStock - SoldQuantity
- Reason for Creation: To calculate the stock quantity remaining after sales.
- Expected Relationship with Target: It may influence product availability and future sales. The relationship depends on the prediction task.
- Potential Business Meaning: Shows the remaining inventory available for sale.
- Potential ML Value: May help inventory forecasting and stock management models.
- Leakage Risk: High if SoldQuantity includes sales that occurred after the prediction time. Use only sales information available before or at the prediction point.
- Final Decision: Retain only when both source values are available at the intended prediction time.

### Final Observation

Subtraction creates useful features by calculating reductions, balances, and remaining values.

Before retaining a feature, verify its business meaning, source data, prediction timing, and leakage risk.

# Topic 4 – Multiplication

## Step 1 – Understand the Concept

Multiplication is a mathematical operation used to calculate the combined value of two or more numerical values.

In Feature Engineering, multiplication helps create new features by combining numerical columns to represent meaningful business information.

Formula:

New Feature = Value A × Value B

When to Use:
- When calculating total amounts from quantity and unit price.
- When combining numerical values to represent business measurements.
- When calculating costs, revenue, or weighted values.

Why to Use:
- To create meaningful features from existing numerical columns.
- To represent relationships between different numerical values.
- To provide additional information for Machine Learning models.

Simple Example:

If one notebook costs ₹50 and a customer buys 4 notebooks:

Total Amount = 50 × 4 = ₹200

Real-World Example:

If a vehicle travels 60 kilometres per hour for 2 hours:

Distance = 60 × 2 = 120 kilometres.

Business Example:

An e-commerce company sells 5 products at ₹800 each.

Total Product Amount = 5 × 800 = ₹4,000.

AI/ML Use Case:

An e-commerce model can use the total purchase amount to understand customer spending behaviour and support purchase prediction.

Multiplication is useful when the numerical values have a meaningful business relationship and compatible units.

# Topic 5 – Division

## Step 1 – Understand the Concept

Division is a mathematical operation used to split one numerical value by another numerical value.

In Feature Engineering, division helps create new features by calculating averages, per-unit values, and proportions.

Formula:

New Feature = Value A / Value B

When to Use:
- When calculating the price per product.
- When finding average values.
- When calculating cost or revenue per unit.
- When comparing two numerical measurements.

Why to Use:
- To understand the value of one quantity relative to another.
- To create meaningful per-unit features.
- To make numerical relationships easier for Machine Learning models to understand.

Simple Example:

If ₹1,000 is shared equally among 5 people:

Amount Per Person = 1000 / 5 = ₹200

Real-World Example:

A vehicle travels 300 kilometres using 10 litres of fuel.

Fuel Efficiency = 300 / 10 = 30 kilometres per litre.

Business Example:

An e-commerce company receives ₹5,000 from selling 10 products.

Revenue Per Product = 5000 / 10 = ₹500.

AI/ML Use Case:

An e-commerce model can use average revenue per product to understand transaction value and customer purchasing behaviour.

Important:

The denominator must not be zero. Both values should have a meaningful relationship and compatible measurement units.

## Step 2 – Demonstrate the Concept

Consider an e-commerce company that records the total product amount and quantity sold in each invoice.

Example Dataset:

| Invoice | TotalAmount | Quantity | TotalCost |
|---------|-------------|----------|-----------|
| INV001  | 1500        | 3        | 900       |
| INV002  | 2400        | 4        | 1600      |
| INV003  | 1000        | 5        | 600       |

Feature 1: AveragePricePerItem

Formula:

AveragePricePerItem = TotalAmount / Quantity

For INV001:

1500 / 3 = ₹500

Feature 2: CostPerItem

Formula:

CostPerItem = TotalCost / Quantity

For INV001:

900 / 3 = ₹300

Why Are These Features Useful?

- AveragePricePerItem represents the average selling amount per product.
- CostPerItem represents the average cost of one product.
- These features help businesses understand product pricing and costs.

AI/ML Application:

- AveragePricePerItem may help analyze customer spending patterns.
- CostPerItem may help understand product cost and profitability.

# Topic 9 – Aggregations

## Step 1 – Understand the Concept

Aggregation is the process of combining multiple numerical values into a single summary value using mathematical operations.

In Feature Engineering, aggregation helps summarize data and create meaningful features such as totals, averages, minimums, and maximums.

Common Aggregation Operations:

- Sum: Calculates the total of numerical values.
- Mean: Calculates the average value.
- Minimum: Finds the smallest value.

- Count: Counts the number of records.

When to Use:

Why to Use:
- To summarize multiple records.
- To identify patterns in customer and business behaviour.
- To reduce detailed transaction data into useful measurements.
- To create meaningful features for Machine Learning models.

Simple Example:

The marks of a student are 70, 80, and 90.

Total Marks = 70 + 80 + 90 = 240

Average Marks = 240 / 3 = 80

Real-World Example:

A person spends ₹100, ₹200, and ₹300 on three shopping trips.

Total Spending = ₹600

Average Spending = ₹200

Business Example:

An e-commerce company calculates the total quantity purchased by each customer to understand their purchasing activity.

AI/ML Use Case:

Customer-level aggregation features such as total spending and average order value can help predict customer behaviour and purchasing patterns.

Important:

Aggregations must use the correct grouping level and only information available at the prediction time. Using future transactions can cause data leakage.

## Step 2 – Demonstrate the Concept

Consider an e-commerce company that records multiple product transactions made by different customers.

Example Dataset:

| CustomerID | Invoice | Quantity | Price |
|------------|---------|----------|-------|
| C101       | INV001  | 2        | 500   |
| C101       | INV002  | 3        | 400   |
| C102       | INV003  | 1        | 800   |
| C102       | INV004  | 4        | 300   |

First, calculate the transaction amount.

TransactionAmount = Quantity × Price

For C101:

- INV001: 2 × 500 = 1000
- INV002: 3 × 400 = 1200

Customer-Level Aggregations:

Feature 1: TotalSales

Formula:

TotalSales = Sum of TransactionAmount

For C101:

1000 + 1200 = 2200

Feature 2: AveragePrice

Formula:

AveragePrice = Mean of Price

For C101:

(500 + 400) / 2 = 450

Feature 3: TotalQuantity

Formula:

TotalQuantity = Sum of Quantity

For C101:

2 + 3 = 5

Why Are These Features Useful?

- TotalSales represents the total spending or sales amount associated with a customer.
- AveragePrice represents the average unit price of purchased products.

AI/ML Application:

These features may help identify high-value customers, understand purchasing behaviour, and support customer segmentation.

```python
import pandas as pd

# Create a sample e-commerce transaction dataset
data = {
    "CustomerID": ["C101", "C101", "C102", "C102"],
    "Invoice": ["INV001", "INV002", "INV003", "INV004"],
    "Quantity": [2, 3, 1, 4],
    "Price": [500, 400, 800, 300]
}

df = pd.DataFrame(data)

# Calculate transaction amount
df["TransactionAmount"] = (
    df["Quantity"] * df["Price"]
)

# Perform customer-level aggregation
customer_features = df.groupby("CustomerID").agg(
    TotalSales=("TransactionAmount", "sum"),
    AveragePrice=("Price", "mean"),
    TotalQuantity=("Quantity", "sum")
).reset_index()

# Display customer-level features
print(customer_features)
```

# Topic 10 – Log Transformation

## Step 1 – Understand the Concept

Log Transformation is a mathematical technique used to reduce the scale of numerical values by applying a logarithmic function.

In Feature Engineering, log transformation is useful when numerical data contains very large values or is highly right-skewed.

Formula:

Log Transformed Value = log(1 + X)

Here:
- X represents the original numerical value.
- log represents the natural logarithm.
- Adding 1 allows zero values to be included.

When to Use:
- When numerical data contains very large values.
- When a feature has a highly right-skewed distribution.
- When the difference between small and large values is very high.
- When reducing the influence of extremely large values.

Why to Use:
- To reduce the scale of large numerical values.
- To reduce right skewness.
- To make numerical distributions easier to analyze.
- To help certain Machine Learning models learn from numerical features.

Simple Example:

Consider the values 10, 100, and 1000.

Applying log(1 + X) reduces the differences between these values.

Real-World Example:

A company's daily website visits may range from a few hundred to several thousand. Log transformation can reduce the scale difference.

Business Example:

An e-commerce company may have customers with very different purchase amounts. Log transformation can help reduce the effect of customers with extremely high spending.

AI/ML Use Case:

Log-transformed purchase amounts may help certain Machine Learning models handle highly skewed transaction values.

Important:

Log transformation requires values greater than or equal to zero when using log(1 + X). It does not automatically remove outliers or guarantee better model performance.

# Topic 14 – Relative Difference

## Step 1 – Understand the Concept

Relative Difference is a numerical feature engineering technique used to measure the difference between two values in relation to a reference value.

It expresses the difference as a proportion or percentage of the reference value.

Formula:

RelativeDifference = (CurrentValue - PreviousValue) / PreviousValue

RelativeDifferencePercentage = RelativeDifference * 100

Where:
- CurrentValue represents the latest numerical value.
- PreviousValue represents the reference or earlier value.

Example:

PreviousPrice = 200
CurrentPrice = 250

RelativeDifference = (250 - 200) / 200

RelativeDifference = 0.25

RelativeDifferencePercentage = 0.25 * 100 = 25%

A positive result indicates an increase, while a negative result indicates a decrease.

This technique helps compare changes across values with different starting amounts.

## Step 2 – Demonstrate the Concept

Simple Example:

Previous Value = 100
Current Value = 120

Relative Difference = (120 - 100) / 100

Relative Difference = 0.20

Percentage Difference = 20%

Real-World Example:

- When measuring percentage growth or decline.
- When comparing price, revenue, sales, or quantity changes over time.

Why to Use:

It provides a relative measure of change rather than only showing the numerical difference.

Important Considerations:

- The reference value must not be zero.
- Both values must use the same unit.
- The reference period and current period must be clearly defined.
- Only information available before the prediction time should be used.

# Topic 15 – Rate Features

## Step 1 – Understand the Concept

Rate Features are numerical features created by measuring how much of an activity happens within a specific amount of time or another reference unit.

A rate represents the amount of change or activity per unit of time or another quantity.

Formula:

Rate = Total Quantity / Time Period

Where:
- Total Quantity represents the amount of activity.
- Time Period represents the duration over which the activity occurred.

Example:

A store sells 100 products in 5 days.

SalesRate = 100 / 5 = 20 products per day

Rate features help represent the speed or frequency of business activities.

They are useful in machine learning when the frequency or speed of an activity may influence the prediction target.

# Topic 16 – Normalized Features

## Step 1 – Understand the Concept

Normalization is a numerical feature engineering technique used to scale numerical values into a common range.

It helps prevent features with larger numerical scales from dominating features with smaller scales in certain machine learning algorithms.

One common method is Min-Max Normalization.

Formula:

X_normalized = (X - X_min) / (X_max - X_min)

Where:
- X represents the original value.
- X_min represents the minimum value.
- X_max represents the maximum value.

Min-Max normalization usually converts values into a range between 0 and 1.

Example:

PurchaseAmount = [100, 200, 300]

Minimum Value = 100
Maximum Value = 300

Normalized Value for 200:

(200 - 100) / (300 - 100) = 0.5

Normalization changes the scale of numerical features without changing their original order.

It is commonly useful for machine learning algorithms that are sensitive to feature scales.

## Step 2 – Demonstrate the Concept

Simple Example:

Consider three values:

[10, 20, 30]

Minimum Value = 10
Maximum Value = 30

Normalized values:

10 = (10 - 10) / (30 - 10) = 0

20 = (20 - 10) / (30 - 10) = 0.5

30 = (30 - 10) / (30 - 10) = 1

Real-World Example:

An online shopping platform may store purchase amounts and product quantities.

Purchase amounts may range from hundreds to thousands, while quantities may range from 1 to 100.

Normalization brings these features into a comparable numerical range.

Business Example:

A retail company may analyze product prices, quantities sold, and customer spending together. Normalization helps prepare these numerical features for algorithms that are sensitive to differences in scale.

AI/ML Use Case:

Normalization can be useful for algorithms such as K-Nearest Neighbors, Support Vector Machines, and neural networks.

When to Use:

- When numerical features have different scales.
- When using distance-based machine learning algorithms.
- When an algorithm is sensitive to feature magnitude.

Why to Use:

It brings numerical features into a common range and helps prevent larger-scale features from dominating calculations.

Important Considerations:

- Min-Max normalization is sensitive to extreme values.
- A constant column where minimum equals maximum cannot be scaled using this formula directly.
- Fit the scaler using training data only.
- Apply the fitted scaler to validation and test data.
- Tree-based models generally do not require normalization.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Create a sample e-commerce dataset
df = pd.DataFrame({
    "CustomerID": [101, 102, 103, 104, 105],
    "PurchaseAmount": [100, 200, 300, 400, 500],
    "PurchaseQuantity": [2, 4, 6, 8, 10]
})

# Select numerical features
features = ["PurchaseAmount", "PurchaseQuantity"]

# Create the Min-Max Scaler
scaler = MinMaxScaler()

# Fit and transform the numerical features
normalized_values = scaler.fit_transform(df[features])

# Create normalized feature columns
normalized_df = pd.DataFrame(
    normalized_values,
    columns=[
        "NormalizedPurchaseAmount",
        "NormalizedPurchaseQuantity"
    ]
)

# Combine original and normalized features
df = pd.concat([df, normalized_df], axis=1)

# Display the updated dataset
print(df)
```
